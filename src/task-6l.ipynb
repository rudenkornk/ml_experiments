**1.** Загрузите выбранные модели из файлов (либо обучите их заново) и оцените качество всех трёх моделей на *одном и том же* тестовом множестве, используя одну и ту же метрику качества.

```python
path = kagglehub.dataset_download("juice0lover/users-vs-bots-classification")
df = pd.read_csv(f"{path}/bots_vs_users.csv")
df["subscribers_count"] = pd.to_numeric(df["subscribers_count"], errors="coerce")
y = df["target"]
df = df.drop(columns=["target", "city"])
```

```python
numeric_cols = df.select_dtypes(include="number").columns
cat_cols = df.select_dtypes(include=["object", "category", "string"]).columns
```

```python
scaler = StandardScaler()
num_imputer = SimpleImputer(strategy="mean", add_indicator=True)
cat_encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("numeric",  make_pipeline(scaler, num_imputer), numeric_cols),
        ("categorical", make_pipeline(cat_encoder), cat_cols),
    ],
).set_output(transform="pandas")
```

```python
model_no_reg = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(C=np.inf, max_iter=1000, intercept_scaling=1000000)),
])
```

```python
model_l1_reg = Pipeline(
    [
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                C=0.5,
                l1_ratio=1.0,
                solver="saga",
                max_iter=1000,
                intercept_scaling=1000000,
            ),
        ),
    ]
)
```

```python
model_l2_reg = Pipeline(
    [
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                C=0.5,
                l1_ratio=0.0,
                solver="newton-cholesky",
                max_iter=1000,
                intercept_scaling=1000000,
            ),
        ),
    ]
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df, y,
    test_size=0.25,
    stratify=y,
    random_state=0,
)
model_no_reg.fit(X_train, y_train)
model_l1_reg.fit(X_train, y_train)
model_l2_reg.fit(X_train, y_train)
print(f"No reg: {model_no_reg.score(X_test, y_test):.2%}")
print(f"L1:     {model_l1_reg.score(X_test, y_test):.2%}")
print(f"L2:     {model_l2_reg.score(X_test, y_test):.2%}")
```

**2.** Теперь проверим различия более формально — с помощью методов AB-тестирования. Рассмотрим сравнение двух моделей между собой. Какие гипотезы следует сформулировать, чтобы проверить, различается ли их качество?

*Указание.* В качестве элементов выборки для каждой модели используйте соответствующие составляющие метрики качества, вычисленные для отдельных объектов:
* $\left(Y_i - \widehat{y}(x_i)\right)^2$ &mdash; для метрики MSE;
* $\left|Y_i - \widehat{y}(x_i)\right|$ &mdash; для метрики MAE;
* $I\left\{Y_i = \widehat{y}(x_i)\right\}$ &mdash; для точности бинарной классификации.

Обозначим истинные вероятности правильной классификации двух моделей через $a_1$ и $a_2$.
Проверим гипотезы:

$$
H_0: a_1 = a_2,
\qquad
H_1: a_1 \ne a_2.
$$

Для каждого объекта тестовой выборки введём индикаторы правильного ответа:

$$
Z_{1i} = \mathbb{I}\{Y_i = \widehat{Y}_{1i}\},
\qquad
Z_{2i} = \mathbb{I}\{Y_i = \widehat{Y}_{2i}\}.
$$

Посчитаем число объектов, на которых правильный ответ дала только первая или только вторая модель:

$$
b = \sum_{i=1}^{n}\mathbb{I}\{Z_{1i}=1,\ Z_{2i}=0\},
\qquad
c = \sum_{i=1}^{n}\mathbb{I}\{Z_{1i}=0,\ Z_{2i}=1\},
\qquad
m = b+c.
$$

Все вероятности далее рассматриваются условно на фиксированном $m>0$.
Для независимых одинаково распределённых тестовых объектов:

$$
B\mid H_0\sim\operatorname{Binom}\left(m,\frac12\right).
$$

Обозначим функцию распределения $\operatorname{Binom}(m,1/2)$ через $F$.
Выделим каждому хвосту уровень значимости $\alpha/2$:

$$
\alpha=0.05.
$$

Критерий:

$$
S=\{b<\ell\}\cup\{b>u\}.
$$

Вероятность ошибки первого рода:

$$
\mathbb{P}(S\mid H_0)
=F(\ell-1)+1-F(u)
\le\frac{\alpha}{2}+\frac{\alpha}{2}
=\alpha.
$$

Ограничения на пороги:

$$
\mathbb{P}(B<\ell\mid H_0)=F(\ell-1)\le\frac{\alpha}{2},
\qquad
\mathbb{P}(B>u\mid H_0)=1-F(u)\le\frac{\alpha}{2}.
$$

Обозначим через $F^{-1}(p)$ наименьшее целое $k$, для которого $F(k)\ge p$.
Пороги можно выразить через квантили:

$$
\ell=F^{-1}\left(\frac{\alpha}{2}\right),
\qquad
u=F^{-1}\left(1-\frac{\alpha}{2}\right).
$$

Итоговый критерий:

$$
S=
\left\{b<F^{-1}\left(\frac{\alpha}{2}\right)\right\}
\cup
\left\{b>F^{-1}\left(1-\frac{\alpha}{2}\right)\right\}.
$$

Проведите все три попарных сравнения моделей на тестовом множестве, используя метод сведения проверки гипотезы к линейной регрессии (см. презентацию). Учтите, что обе модели в каждой паре оцениваются на одних и тех же объектах. Используйте уровень значимости 5%.

Для каких пар моделей удалось обнаружить статистически значимые различия в качестве?

```python
import numpy as np
import pandas as pd
from scipy.stats import binom

alpha = 0.05

actual = np.asarray(y_test)

correct_no_reg = np.asarray(model_no_reg.predict(X_test)) == actual
correct_l1 = np.asarray(model_l1_reg.predict(X_test)) == actual
correct_l2 = np.asarray(model_l2_reg.predict(X_test)) == actual


def compare_models(correct_first, correct_second):
    b = int(np.sum(correct_first & ~correct_second))
    c = int(np.sum(~correct_first & correct_second))
    m = b + c

    lower = int(binom.ppf(alpha / 2, n=m, p=0.5))
    upper = int(binom.ppf(1 - alpha / 2, n=m, p=0.5))

    return {
        "b": b,
        "lower": lower,
        "upper": upper,
        "Отвергаем $H_0$": b < lower or b > upper,
    }


comparisons = pd.DataFrame.from_dict(
    {
        "Без регуляризации vs L1": compare_models(correct_no_reg, correct_l1),
        "Без регуляризации vs L2": compare_models(correct_no_reg, correct_l2),
        "L1 vs L2": compare_models(correct_l1, correct_l2),
    },
    orient="index",
)

display(comparisons)
```

Модели статистически значимо не отличаются.
