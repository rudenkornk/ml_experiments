Рассмотрим задачу бинарной классификации, причем $\mathscr{Y} = \{+1, -1\}$. Пусть так же $\widehat{y}$ &mdash; некоторый классификатор, предсказывающий <u><i>степень принадлежности</i></u> классу. При этом решающее правило имеет вид
 $f(x) = \text{sign}\left(\widehat{y}(x)\right)$.
Исследуем логистическую функцию потерь:
$$\mathcal{L}(y, z) = \log \big(1 + \exp(-yz) \big).$$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 500)
loss = np.logaddexp(0, -x)

plt.plot(x, loss)
plt.xlabel("x")
plt.ylabel("L(x)")
plt.title("L(x) = log(1 + exp(-x))")
plt.grid(True)
plt.show()
```

**1.** Выпишите для градиентного спуска формулу для вектора сдвига (градиент предсказаний). Сформулируйте задачу поиска новой базовой модели $b_t$ по их приближению.

### 1.

$$
\mathcal{L}(y,z)=\log\left(1+e^{-yz}\right).
$$

$$
\frac{\partial \mathcal{L}}{\partial z}
=
-\frac{y e^{-yz}}{1+e^{-yz}},
\qquad
Q(\hat{y})=\sum_{i=1}^{n}\mathcal{L}(Y_i,\hat{y}_i).
$$

Следовательно, вектор градиента:

$$
\nabla_{\hat{y}}Q
=
\begin{pmatrix}
\vdots\\
\dfrac{\partial \mathcal{L}(Y_i,\hat{y}_i)}
      {\partial \hat{y}_i}\\
\vdots
\end{pmatrix}
=
\begin{pmatrix}
\vdots\\
-\dfrac{Y_i e^{-Y_i\hat{y}_i}}
       {1+e^{-Y_i\hat{y}_i}}\\
\vdots
\end{pmatrix}.
$$

Задача поиска $b_t$ — предсказывать
$\nabla_{\hat{y}}Q$ по признакам $X_i$, где
$\hat{y}_i$ — выход предыдущей итерации.

Отдельно ищется коэффициент $\tilde{\gamma}$:

$$
\tilde{\gamma}
=
\operatorname*{arg\,min}_{\gamma}
Q\left(\hat{y}+\gamma b_t\right).
$$

**2.** Предположим, модель градиентного бустинга $\widehat{y}_{t - 1}$ уже построена. Выпишите вид функционала ошибки $Q(\widehat{y}_t) = \sum\limits_{i = 1}^n \mathcal{L}\left(Y_i, \widehat{y}_t(x_i)\right)$ для логистической функции потерь. Одинаковый ли вклад вносят разные объекты в ошибку при построении новой базовой модели $b_t$?



### 2.

$$
\begin{aligned}
Q(\hat{y}_t)
&=\sum_{i=1}^{n}\mathcal{L}(Y_i,\hat{y}_{i,t})\\
&=\sum_{i=1}^{n}
\mathcal{L}\left(
Y_i,\hat{y}_{i,t-1}+\tilde{\gamma}_t b_t(X_i)
\right)\\
&=\sum_{i=1}^{n}
\log\left(
1+e^{-Y_i(\hat{y}_{i,t-1}+\tilde{\gamma}_t b_t(X_i))}
\right)\\
&=\sum_{i=1}^{n}
\log\left(
1+e^{-Y_i\hat{y}_{i,t-1}}
  e^{-Y_i\tilde{\gamma}_t b_t(X_i)}
\right).
\end{aligned}
$$

Вклад разных объектов различается и зависит от
$Y_i\hat{y}_{i,t-1}$.

Если $Y_i$ и $\hat{y}_{i,t-1}$ имеют разные знаки,
то вклад примерно линеен по $-Y_i\hat{y}_{i,t-1}$.

**3.** Посмотрите на формулу для вектора сдвигов, полученную ранее. Как она выражается через таргеты на объектах обучающей выборки? Одинаковый ли вклад вносят разные объекты в формирование вектора сдвигов при построении новой базовой модели $b_t$?


### 3.

$$
\left(
\nabla_{\hat{y}}Q(\hat{y}_{t-1})
\right)_i
=
-\frac{Y_i e^{-Y_i\hat{y}_{i,t-1}}}
       {1+e^{-Y_i\hat{y}_{i,t-1}}}
=
Y_i w_i,
$$

где

$$
w_i
=
-\frac{\exp(-Y_i\hat{y}_{i,t-1})}
       {1+\exp(-Y_i\hat{y}_{i,t-1})}.
$$

То есть в некотором смысле новый таргет —
это исходный $Y_i$, умноженный на некоторый вес $w_i$.

Соответственно, вклад объектов также различается
в зависимости от $w_i$.

**4.**  На лекции было показано, что для экспоненциальной функции есть проблема: базовый классификатор может настраиваться только на шумовые объекты. Наблюдается ли такая проблема у логистической функции потерь? Как она ведет себя на объектах с сильно отрицательным, околонулевым, сильно положительным отсупами?

### 4.

У логистической функции потерь такой проблемы нет.

Величина $|w_i|<1$, поэтому модуль компоненты
градиента не превышает $|Y_i|$. Соответственно,
доминирующего влияния случайных выбросов нет.

Если обозначить отступ через
$m_i=Y_i\hat{y}_i$, то

$$
\mathcal{L}(m_i)=\log\left(1+e^{-m_i}\right).
$$

Как видно из графика, для больших положительных
отступов функция потерь стремится к нулю:

$$
\mathcal{L}(m)\to 0
\qquad (m\to+\infty).
$$

При нулевом отступе:

$$
\mathcal{L}(0)=\log 2.
$$

Для больших по модулю отрицательных отступов
функция потерь возрастает примерно линейно:

$$
\mathcal{L}(m)\sim -m
\qquad (m\to-\infty).
$$
