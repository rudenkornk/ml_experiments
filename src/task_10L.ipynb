```python
from pathlib import Path
from tempfile import TemporaryDirectory
from urllib.request import urlretrieve
from zipfile import ZipFile

import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

url = "https://thetahat.ru/files/ad/main/7/cats.zip"

with TemporaryDirectory() as tmp:
    zip_path = Path(tmp) / "cats.zip"
    urlretrieve(url, zip_path)

    with ZipFile(zip_path) as archive:
        names = sorted(name for name in archive.namelist() if name.lower().endswith((".jpg", ".jpeg", ".png")))
        cats = []
        for name in names:
            with archive.open(name) as file, Image.open(file) as img:
                rgb = img.convert("RGB")
                if rgb.size != (64, 64):
                    raise ValueError(f"{name}: expected 64×64, got {rgb.size}")
                cats.append(np.asarray(rgb, dtype=np.float32) / 255)

cats = np.stack(cats).reshape(len(cats), 64 * 64 * 3)
print(f"Размерность cats: {cats.shape}")
```

#### 1. Свойства метрики в пространстве котиков

Как мы знаем из лекции, в пространствах большой размерности расстояния между случайными объектами становятся неинформативными. Эта проблема известна как *проклятие размерности*, и она влечет соответствующие ограничения на использование методов, основанных на использовании расстоянии между объектами.

Прежде всего давайте исследуем, наблюдается ли проблема проклятия размерности в пространстве котиков. Для этого выберите не менее 10000 случайных пар изображений и посчитайте расстояния между ними. Нормируйте все расстояния, поделив на среднее полученных расстояний. Визуализируйте гистограмму или KDE-оценку плотности нормированных расстояний.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
distances = np.empty(10_000)

for k in range(len(distances)):
    i, j = rng.choice(len(cats), size=2, replace=False)
    distances[k] = np.linalg.norm(cats[i] - cats[j])

normalized = distances / distances.mean()

plt.figure(figsize=(8, 4))
plt.hist(normalized, bins=50, density=True, alpha=0.8)
plt.axvline(1, color="black", linestyle="--", label="Среднее")
plt.xlabel("Расстояние / среднее расстояние")
plt.ylabel("Плотность")
plt.title("Расстояния между случайными парами изображений котов")
plt.legend()
plt.show()
```

Повторите те же действия к результату применения PCA, рассмотрев разное количество компонент, например, для 5, 10, 30, 100 и 500. Обратите внимание, что PCA нужно обучать на всех изображениях, а не только для тех, для которых вы будете считать расстояния. Примеры обучения PCA можно посмотреть в ноутбуке с занятия.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

components = [5, 10, 30, 100, 500]
pca = PCA(n_components=max(components), svd_solver="randomized", random_state=42)
cats_pca = pca.fit_transform(cats)

rng = np.random.default_rng(42)
i, j = np.array([rng.choice(len(cats), 2, replace=False) for _ in range(10_000)]).T

normalized_by_k = {}
for k in components:
    distances = np.linalg.norm(cats_pca[i, :k] - cats_pca[j, :k], axis=1)
    normalized_by_k[k] = distances / distances.mean()
```

```python
np.mean(np.abs((normalized_by_k[500] - normalized)/normalized))
```

Сравните результаты и сделайте выводы

```python
bins = np.histogram_bin_edges(np.concatenate(list(normalized_by_k.values())), bins=60)
plt.figure(figsize=(10, 5))
for k, values in normalized_by_k.items():
    plt.hist(values, bins=bins, density=True, histtype="step",
             linewidth=1.6, label=f"k = {k}, σ = {values.std():.2f}")

plt.axvline(1, color="gray", linestyle="--")
plt.xlabel("Расстояние / среднее расстояние")
plt.ylabel("Плотность")
plt.title("Распределение расстояний после PCA")
plt.legend()
plt.show()
```

При меньших k выходит что дисперсия больше.  
То есть при меньшей размерности пространства (выбранного в главных компонентах) разброс расстояний между случайными парами выше.  
При увеличении размерности они начинают стягиваться ближе к одной точке и выходит что расстояние начинает нести меньше информации.  
Правда в этом случае это как будто не очень заметно. Для k=500 дисперсия меньше всего вдвое, что все еще относительно неплохо.  

#### 2. Кластеризация котиков 🙀

На занятии мы рассмотрели кластеризацию изображений котиков на 3 кластера. По примерам кажется, что кластеры соответствуют тёмным, промежуточным и светлым изображениям. Подтверждается ли это на всём датасете?

Для ответа на вопрос для каждой фотографии вычислите среднюю яркость (среднее по всем пикселям и каналам), затем визуализируйте распределения яркости внутри кластеров. Найдите несколько изображений, которые плохо соответствуют описанию своего кластера.

```python
pca = PCA(n_components=100)
cats_projection = pca.fit_transform(cats.astype(np.float32))

print(f"Размерность данных для кластеризации: {cats_projection.shape}")
print(f"Доля объясненной дисперсии: {pca.explained_variance_ratio_.sum():.1%}")

n_clusters_grid = range(3, 11)
silhouette_scores = []
kmeans_models = {}

for n_clusters in tqdm(n_clusters_grid):
    kmeans = KMeans(n_clusters=n_clusters, n_init=10, random_state=RANDOM_STATE)
    kmeans.fit(cats_projection)
    kmeans_models[n_clusters] = kmeans

    score = silhouette_score(
        cats_projection,
        kmeans.labels_,
        sample_size=3000,
        random_state=RANDOM_STATE,
    )
    silhouette_scores.append(score)
```

```python
labels = kmeans_models[3].labels_
brightness = cats.mean(axis=1)

groups = sorted(np.unique(labels), key=lambda g: brightness[labels == g].mean())
names = ["Тёмный", "Промежуточный", "Светлый"]
bins = np.histogram_bin_edges(brightness, bins=40)

plt.figure(figsize=(9, 4))
for g, name in zip(groups, names):
    values = brightness[labels == g]
    plt.hist(values, bins=bins, density=True, histtype="step", linewidth=2,
             label=f"{name} (№{g}): среднее {values.mean():.2f}")

plt.xlabel("Средняя яркость фотографии")
plt.ylabel("Плотность")
plt.legend()
plt.show()
```

```python
# Исключения.
fig, axes = plt.subplots(3, 3, figsize=(7, 7), layout="constrained")

for row, (g, name) in enumerate(zip(groups, names)):
    indices = np.flatnonzero(labels == g)
    values = brightness[indices]

    if row == 0:
        score = values
    elif row == 1:
        score = np.abs(values - np.median(values))
    else:
        score = -values

    examples = indices[np.argsort(score)[-3:][::-1]]
    for ax, i in zip(axes[row], examples):
        ax.imshow(cats[i].reshape(64, 64, 3))
        ax.set_title(f"{name} · №{i}\nяркость {brightness[i]:.2f}")
        ax.axis("off")

plt.show()
```

Самые светлые в темном кластере и самые темные в светлом кластере все еще достаточно подходят своим кластерам.

Для каждого кластера найдите по пять изображений, ближайших к его центру, и по пять наиболее далеких среди изображений, принадлежащих этому кластеру. Расстояния вычисляйте в пространстве компонент PCA, в котором обучался k-means. Что объединяет эти изображения и чем отличаются далекие?

```python
import numpy as np
import matplotlib.pyplot as plt

model = kmeans_models[3]
labels = model.labels_
distances = model.transform(cats_projection)

for cluster in range(model.n_clusters):
    members = np.flatnonzero(labels == cluster)
    ordered = members[np.argsort(distances[members, cluster])]

    fig, axes = plt.subplots(2, 5, figsize=(10, 4.5), layout="constrained")
    fig.suptitle(f"Кластер {cluster}: сверху — ближайшие, снизу — самые далёкие")

    for row, indices in zip(axes, [ordered[:5], ordered[-5:][::-1]]):
        for ax, i in zip(row, indices):
            ax.imshow(cats[i].reshape(64, 64, 3))
            ax.set_title(f"№{i} · d={distances[i, cluster]:.1f}")
            ax.axis("off")

    plt.show()
```

Ближайшие к центру хорошо описывают ожидаемо среднее значение яркости каждого кластера.  
В темном кластере ближайшие к центру темные (но не самые темные) коты.  
Далекие коты начинают содержать признаки чужого кластера.  Больше темных участков в светлом кластере, больше светлых в темном кластере.  
В среднем кластере самые далекие содержат почти черно-белые изображения.

Изменились ли кластеры? Какие свойства изображений теперь объединяют котов внутри кластера? Учтите, что номера кластеров не фиксированы, они могут меняться.

```python
from sklearn.base import clone

cats_centered = cats - cats.mean(axis=1, keepdims=True)

pca_centered = clone(pca)
projection_centered = pca_centered.fit_transform(cats_centered)

model_centered = clone(kmeans_models[3])
labels_centered = model_centered.fit_predict(projection_centered)
distances_centered = model_centered.transform(projection_centered)
```

```python
for g in range(model_centered.n_clusters):
    members = np.flatnonzero(labels_centered == g)
    ordered = members[np.argsort(distances_centered[members, g])]

    fig, axes = plt.subplots(2, 5, figsize=(10, 4.5), layout="constrained")
    fig.suptitle(f"Кластер {g} после центрирования: сверху — ближайшие, снизу — самые далёкие")

    for row, indices in zip(axes, [ordered[:5], ordered[-5:][::-1]]):
        for ax, i in zip(row, indices):
            ax.imshow(cats[i].reshape(64, 64, 3))
            ax.set_title(f"№{i} · d={distances_centered[i, g]:.1f}")
            ax.axis("off")

    plt.show()
```

```python
contrast = cats_centered.std(axis=1, keepdims=True)
cats_normalized = cats_centered / np.where(contrast > 0, contrast, 1.0)

pca_normalized = clone(pca_centered)
projection_normalized = pca_normalized.fit_transform(cats_normalized)

model_normalized = clone(model_centered)
labels_normalized = model_normalized.fit_predict(projection_normalized)
distances_normalized = model_normalized.transform(projection_normalized)
```

```python
for g in range(model_normalized.n_clusters):
    members = np.flatnonzero(labels_normalized == g)
    ordered = members[np.argsort(distances_normalized[members, g])]

    fig, axes = plt.subplots(2, 5, figsize=(10, 4.5), layout="constrained")
    fig.suptitle(f"Кластер {g} после нормировки контраста")

    for row, indices in zip(axes, [ordered[:5], ordered[-5:][::-1]]):
        for ax, i in zip(row, indices):
            ax.imshow(cats[i].reshape(64, 64, 3))
            ax.set_title(f"№{i} · d={distances_normalized[i, g]:.1f}")
            ax.axis("off")

    plt.show()
```

```python
area_columns = [column for column in data.columns if column.startswith("sqft_")]
discrete_columns = ["bedrooms", "floors", "waterfront", "view", "condition", "grade"]
zero_count_labels = {"yr_renovated": "без ремонта", "sqft_basement": "0", "view": "0"}
plot_columns = [
    "date",
    "price",
    *area_columns,
    "bedrooms",
    "bathrooms",
    *discrete_columns[1:],
    "yr_built",
    "yr_renovated",
]

n_cols = 4
n_rows = (len(plot_columns) + n_cols - 1) // n_cols
LOT_THRESHOLD = 60_000
lot_groups = [
    (f"sqft_lot ≤ {LOT_THRESHOLD:,}", data_train.loc[data_train["sqft_lot"].le(LOT_THRESHOLD)]),
    (f"sqft_lot > {LOT_THRESHOLD:,}", data_train.loc[data_train["sqft_lot"].gt(LOT_THRESHOLD)]),
]
panels = []
for label, group in lot_groups:
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(17, 3 * n_rows), layout="constrained")
    fig.suptitle(f"{label} · n = {len(group):,}")
    for ax in axes.ravel()[len(plot_columns) :]:
        ax.set_visible(False)
    panels.append(axes.ravel())

for index, column in enumerate(plot_columns):
    values = data_train[column].dropna()
    if column == "date":
        values = pd.to_datetime(values, format="%Y%m%dT%H%M%S").dt.to_period("M")
    zero_label = zero_count_labels.get(column)
    plotted_values = values[values.ne(0)] if zero_label is not None else values
    is_categorical = column == "date" or column in discrete_columns
    use_group_bins = column in {"price", "sqft_lot"}
    categories = pd.Index(plotted_values.unique()).sort_values() if is_categorical else None
    bins = None
    if column == "bathrooms":
        bins = np.arange(0, int(values.max()) + 2)
    elif not is_categorical and not use_group_bins:
        bins = np.histogram_bin_edges(plotted_values, bins=30 if column == "yr_renovated" else 40)

    for (_, group), panel_axes in zip(lot_groups, panels, strict=True):
        ax = panel_axes[index]
        group_values = values.loc[values.index.isin(group.index)]
        ax.set(title=column, ylabel="Число продаж")
        if zero_label is not None:
            zeros = group_values.eq(0)
            ax.set_title(f"{column} ({zero_label}: {zeros.sum():,})")
            group_values = group_values[~zeros]

        if categories is not None:
            counts = group_values.value_counts().reindex(categories, fill_value=0)
            ax.bar(categories.astype(str), counts.to_numpy())
            if column == "date":
                ax.tick_params(axis="x", labelrotation=60)
        elif use_group_bins:
            ax.hist(group_values, bins=40)
        elif bins is not None:
            ax.hist(group_values, bins=bins)
            if column == "bathrooms":
                ax.set_xticks(bins)

plt.show()
```

```python
# Поработаем с признаками.

from sklearn.preprocessing import FunctionTransformer

def add_house_features(X):
    """Добавим минимум разумных признаков.
        * Есть ли подвал.
        * Есть ли оценка вида.
        * Дата продажи в днях.
        * sin/cos по годовому периоду.
    """
    sale_date = pd.to_datetime(X["date"], format="%Y%m%dT%H%M%S")
    days_in_year = 365 + sale_date.dt.is_leap_year.astype(int)
    sale_angle = 2 * np.pi * (sale_date.dt.dayofyear - 1) / days_in_year
    return X.assign(
        has_basement=X["sqft_basement"].gt(0).astype("int8"),
        has_view=X["view"].gt(0).astype("int8"),
        has_renovation=X["yr_renovated"].gt(0).astype("int8"),
        sale_days=(sale_date - pd.Timestamp("2015-01-01")).dt.days,
        sale_day_sin=np.sin(sale_angle),
        sale_day_cos=np.cos(sale_angle),
    )

house_features = FunctionTransformer(add_house_features)
```

```python
import geopandas as gpd

lake_crs = "EPSG:32610"
shoreline = lakes.to_crs(lake_crs).geometry.union_all().boundary

def add_lake_distance(X):
    """Добавим еще расстояние до берега озера.
       По карте как будто это имеет значение.    
    """
    points = gpd.GeoSeries(
        gpd.points_from_xy(X["long"], X["lat"]),
        index=X.index,
        crs="EPSG:4326",
    ).to_crs(lake_crs)

    return X.assign(lake_distance_m=points.distance(shoreline))

lake_features = FunctionTransformer(add_lake_distance)
```

```python
from catboost import CatBoostRegressor
from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.preprocessing import OneHotEncoder, TargetEncoder
from xgboost import XGBRegressor
from sklearn.model_selection import KFold, ParameterSampler, cross_val_score
import optuna
```

```python
def build_pipeline(
    feature_steps: tuple[FunctionTransformer, ...],
    category_step: FunctionTransformer | ColumnTransformer,
    model: XGBRegressor | CatBoostRegressor | GradientBoostingRegressor,
) -> Pipeline:
    """Combine feature generation, category preprocessing, and an estimator."""
    return make_pipeline(
        *(clone(step) for step in feature_steps),
        FunctionTransformer(pd.DataFrame.drop, kw_args={"columns": ["id", "date"], "errors": "ignore"}),
        category_step,
        model,
    )


zip_only = ["zipcode"]
extra_categories = [*zip_only, "view", "condition", "grade", "waterfront"]
engineered_features = (house_features, lake_features)
xgb_category_types = {
    column: pd.CategoricalDtype(categories=sorted(X_train[column].dropna().unique())) for column in extra_categories
}

feature_specs = {
    "Исходные": ((), zip_only),
    "Новые": (engineered_features, zip_only),
    "Новые + категории": (engineered_features, extra_categories),
}

model_specs = {
    "XGBoost": (
        lambda columns: FunctionTransformer(
            pd.DataFrame.astype,
            kw_args={"dtype": {column: xgb_category_types[column] for column in columns}},
        ),
        lambda _columns: XGBRegressor(
            tree_method="hist",
            enable_categorical=True,
            random_state=RANDOM_STATE,
        ),
    ),
    "CatBoost": (
        lambda columns: FunctionTransformer(pd.DataFrame.astype, kw_args={"dtype": dict.fromkeys(columns, str)}),
        lambda columns: CatBoostRegressor(
            cat_features=columns,
            random_seed=RANDOM_STATE,
            verbose=False,
            allow_writing_files=False,
        ),
    ),
    "sklearn GB": (
        lambda columns: ColumnTransformer(
            [("one_hot", OneHotEncoder(handle_unknown="ignore"), columns)],
            remainder="passthrough",
        ),
        lambda _columns: GradientBoostingRegressor(random_state=RANDOM_STATE),
    ),
}
```

```python
boosting_pipelines: dict[tuple[str, str], Pipeline] = {}
scores: list[dict[str, str | float]] = []

for model_name, (make_category_step, make_model) in tqdm(model_specs.items(), desc="Модели"):
    for variant, (feature_steps, category_columns) in tqdm(feature_specs.items(), desc=model_name, leave=False):
        pipeline = build_pipeline(
            feature_steps,
            make_category_step(category_columns),
            make_model(category_columns),
        )
        pipeline.fit(X_train, y_train)
        pred = pipeline.predict(X_test)

        boosting_pipelines[model_name, variant] = pipeline
        scores.append(
            {
                "Модель": model_name,
                "Вариант": variant,
                "MAPE, %": 100 * mean_absolute_percentage_error(y_test, pred),
            }
        )

boosting_mape = (
    pd.DataFrame(scores)
    .pivot_table(index="Модель", columns="Вариант", values="MAPE, %")
    .reindex(index=list(model_specs), columns=list(feature_specs))
)
display(boosting_mape.round(3))
```

```python
from sklearn.metrics.pairwise import haversine_distances

def add_extra_features(X: pd.DataFrame) -> pd.DataFrame:
    """Добавим несклько фичей.
        * Сравнение площадей с соседями.
        * Сравнение площади подвала и жилой площади.
        * Возраст постройки.
        * Время со дня ремонта.
        * Расстояние до центра Сиэттла и Бэллвью.    
    """
    sale_year = pd.to_datetime(X["date"], format="%Y%m%dT%H%M%S").dt.year
    renovation_year = X["yr_renovated"].where(X["yr_renovated"] > 0)

    houses_rad = np.deg2rad(X[["lat", "long"]].to_numpy())
    centers_rad = np.deg2rad([
        [47.6062, -122.3321],  # Центр Сиэттла.
        [47.6101, -122.2015],  # Центр Бэллвью.
    ])
    distances_km = 6371.0 * haversine_distances(houses_rad, centers_rad)

    return X.assign(
        living_to_living15=X["sqft_living"] / X["sqft_living15"].replace(0, np.nan),
        lot_to_lot15=X["sqft_lot"] / X["sqft_lot15"].replace(0, np.nan),
        basement_share=X["sqft_basement"] / X["sqft_living"].replace(0, np.nan),
        age_at_sale=sale_year - X["yr_built"],
        years_since_renovation=sale_year - renovation_year,
        distance_to_seattle_km=distances_km[:, 0],
        distance_to_bellevue_km=distances_km[:, 1],
    )
```

```python
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.neighbors import BallTree


def add_layout_features(X):
    """Добавление фичей работает. Ну, добавим тогда еще.
        * Площадь одной спальни.
        * Сколько ванн приходится на одну спальню.
        * Процент жилой площади над землей на один этаж.
    """
    bedrooms = X["bedrooms"].replace(0, np.nan)
    floors = X["floors"].replace(0, np.nan)
    lot = X["sqft_lot"].replace(0, np.nan)

    return X.assign(
        living_per_bedroom=X["sqft_living"] / bedrooms,
        bathrooms_per_bedroom=X["bathrooms"] / bedrooms,
        estimated_lot_coverage=(X["sqft_above"] / floors) / lot,
    )
```

```python
class NeighborhoodFeatures(TransformerMixin, BaseEstimator):
    """
    Попробуем начать сравнивать с соседями.
    Добавим оценку ближайших соседей и сравнение оценки дома с соседями.
    """
    def __init__(self, n_neighbors=20):
        self.n_neighbors = n_neighbors

    def fit(self, X, y=None):
        homes = X.drop_duplicates("id")
        self.ids_ = homes["id"].to_numpy()
        self.grades_ = homes["grade"].to_numpy()
        self.years_ = homes["yr_built"].to_numpy()

        coords = np.deg2rad(homes[["lat", "long"]].to_numpy())
        self.tree_ = BallTree(coords, metric="haversine")
        return self

    def transform(self, X):
        coords = np.deg2rad(X[["lat", "long"]].to_numpy())
        distances, indices = self.tree_.query(coords, k=self.n_neighbors + 1)

        median_grade = np.empty(len(X))
        median_year = np.empty(len(X))
        radius_km = np.empty(len(X))

        for i, house_id in enumerate(X["id"]):
            # Исключаем собственный дом, даже если это другая его продажа.
            other = self.ids_[indices[i]] != house_id
            neighbors = indices[i][other][:self.n_neighbors]
            neighbor_distances = distances[i][other][:self.n_neighbors]

            median_grade[i] = np.median(self.grades_[neighbors])
            median_year[i] = np.median(self.years_[neighbors])
            radius_km[i] = 6371.0 * neighbor_distances[-1]

        return X.assign(
            neighbor_grade_median=median_grade,
            grade_vs_neighbors=X["grade"] - median_grade,
            neighbor_radius_km=radius_km,
            neighbor_year_median=median_year,
            year_vs_neighbors=X["yr_built"] - median_year,
        )
```

```python
NEIGHBORS = 20
N_FOLDS = 3
LOCAL_COLUMNS = ["local_price_per_sqft_lot", "local_price_per_sqft_living"]


def ridge_inputs(homes: pd.DataFrame) -> np.ndarray:
    """Return the information available about the sale before knowing its price."""
    dates = pd.to_datetime(homes["date"], format="%Y%m%dT%H%M%S")
    renovated = homes["yr_renovated"].gt(0)
    return np.column_stack(
        (
            (dates - pd.Timestamp("2015-01-01")).dt.days,
            dates.dt.year - homes["yr_built"],
            (dates.dt.year - homes["yr_renovated"]).where(renovated, 0),
            renovated,
        )
    ).astype(float)


def local_price_features(reference: pd.DataFrame, prices: pd.Series, query: pd.DataFrame, *, desc="Local Ridge") -> pd.DataFrame:
    """Fit a two-output Ridge per query using all reference sales of its nearest other homes."""
    homes = reference.drop_duplicates("id")
    ids = homes["id"].to_numpy()
    tree = BallTree(np.deg2rad(homes[["lat", "long"]].to_numpy()), metric="haversine")
    candidates = tree.query(
        np.deg2rad(query[["lat", "long"]].to_numpy()), k=min(NEIGHBORS + 1, len(homes)), return_distance=False
    )
    sales_by_id = reference.groupby("id").indices
    inputs, query_inputs = ridge_inputs(reference), ridge_inputs(query)
    targets = prices.to_numpy()[:, None] / reference[["sqft_lot", "sqft_living"]].to_numpy()
    predictions = np.full((len(query), len(LOCAL_COLUMNS)), np.nan)

    queries = zip(query["id"], candidates, strict=True)
    for row, (own_id, indices) in enumerate(tqdm(queries, total=len(query), desc=desc, unit="дом")):
        neighbors = ids[indices]
        neighbors = neighbors[neighbors != own_id][:NEIGHBORS]
        if len(neighbors):
            rows = np.concatenate([sales_by_id[neighbor] for neighbor in neighbors])
            model = make_pipeline(StandardScaler(), Ridge(alpha=1.0))
            model.fit(inputs[rows], targets[rows])
            predictions[row] = model.predict(query_inputs[row : row + 1])[0]

    return pd.DataFrame(predictions, index=query.index, columns=LOCAL_COLUMNS)


def oof_local_prices(homes: pd.DataFrame, prices: pd.Series, cv: KFold) -> pd.DataFrame:
    """Generate local Ridge features for each row from the other inner folds."""
    result = pd.DataFrame(np.nan, index=homes.index, columns=LOCAL_COLUMNS)
    for fold, (ref, query) in enumerate(cv.split(homes), start=1):
        result.iloc[query] = local_price_features(
            homes.iloc[ref], prices.iloc[ref], homes.iloc[query], desc=f"Ridge OOF {fold}/{cv.get_n_splits()}"
        ).to_numpy()
    return result
```

```python
from sklearn.utils.validation import check_is_fitted

NEIGHBORS = 20
N_FOLDS = 3

class LocalRidgeFeatures(TransformerMixin, BaseEstimator):

    def __init__(self, cv: KFold) -> None:
        """Set the folds used to generate training features."""
        self.cv = cv

    def fit(self, homes: pd.DataFrame, y: pd.Series) -> LocalRidgeFeatures:
        self.reference_ = homes.copy()
        self.prices_ = y.copy()
        return self

    def fit_transform(self, X: pd.DataFrame, y: pd.Series | None = None, **fit_params: object) -> pd.DataFrame:
        if y is None:
            msg = "Local Ridge requires training prices."
            raise ValueError(msg)
        self.fit(X, y, **fit_params)
        local = oof_local_prices(X, y, self.cv)
        return pd.concat([X, local], axis=1)

    def transform(self, homes: pd.DataFrame) -> pd.DataFrame:
        check_is_fitted(self, ["reference_", "prices_"])
        local = local_price_features(self.reference_, self.prices_, homes, desc="Ridge · predict")
        return pd.concat([homes, local], axis=1)

x_holdout = data_test.drop(columns="price")
inner_cv = KFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)

my_best_model_3 = make_pipeline(
    LocalRidgeFeatures(cv=inner_cv),
    FunctionTransformer(add_layout_features),
    NeighborhoodFeatures(n_neighbors=NEIGHBORS),
    FunctionTransformer(add_house_features),
    FunctionTransformer(add_lake_distance),
    FunctionTransformer(add_extra_features),
    FunctionTransformer(pd.DataFrame.drop, kw_args={"columns": ["id", "date"]}),
    FunctionTransformer(pd.DataFrame.astype, kw_args={"dtype": {"zipcode": str}}),
    ColumnTransformer(
        [("zip_mean", TargetEncoder(target_type="continuous", smooth="auto", cv=inner_cv), ["zipcode"])],
        remainder="passthrough",
        verbose_feature_names_out=False,
    ).set_output(transform="pandas"),
    CatBoostRegressor(
        **study.best_params,
        random_seed=RANDOM_STATE,
        verbose=250,
        allow_writing_files=False,
    ),
)
```

```python
def refinement_objective(trial: optuna.Trial) -> float:
    """Return mean validation MAPE in percent across the three fixed folds."""
    depth = starting_params["depth"]
    iterations = starting_params["iterations"]
    rate = starting_params["learning_rate"]
    l2 = starting_params["l2_leaf_reg"]
    params = {
        "depth": trial.suggest_int("depth", max(1, depth - 2), min(16, depth + 2)),
        "iterations": trial.suggest_int("iterations", max(250, iterations - 1000), iterations + 1000, step=250),
        "learning_rate": trial.suggest_float("learning_rate", rate / 2, rate * 2, log=True),
        "l2_leaf_reg": trial.suggest_float("l2_leaf_reg", l2 / 2, l2 * 2, log=True),
    }
    fold_scores = []
    for fit_features, fit_prices, valid_features, valid_prices in prepared_folds:
        model = CatBoostRegressor(**(current_params | params | {"verbose": False}))
        model.fit(fit_features, fit_prices)
        fold_scores.append(mean_absolute_percentage_error(valid_prices, model.predict(valid_features)))
    return float(100 * np.mean(fold_scores))
```

```python
EARTH_RADIUS_KM = 6371.0
LAKE_CRS = "EPSG:32610"
CITY_CENTERS = np.deg2rad([[47.6062, -122.3321], [47.6101, -122.2015]])
PRICE_COLUMNS = [
    "neighbor_price_per_sqft_living", "neighbor_price_per_sqft_lot",
    "neighbor_price_per_sqft_living_relative_iqr", "comparable_price_per_sqft_living",
]


class _NeighborPrices:
    def __init__(self, x: pd.DataFrame, y: np.ndarray, n_neighbors: int) -> None:
        self.n_neighbors = n_neighbors
        homes = x.drop_duplicates("id")
        self.ids = homes["id"].to_numpy()
        self.tree = BallTree(np.deg2rad(homes[["lat", "long"]].to_numpy()), metric="haversine")
        areas = x[["sqft_living", "sqft_lot"]].replace(0, np.nan)
        ratios = areas.rdiv(y, axis="index")
        self.prices = ratios.groupby(x["id"]).median().loc[self.ids].to_numpy()
        self.sale_prices = ratios["sqft_living"].to_numpy()
        self.sales = self._comparison_features(x)
        scale = self.sales.std(axis=0)
        self.scale = np.where(scale > 0, scale, 1.0)
        by_id = x.groupby("id").indices
        self.first_sale = np.array([by_id[home_id][0] for home_id in self.ids])
        self.repeats = {i: by_id[home_id] for i, home_id in enumerate(self.ids) if len(by_id[home_id]) > 1}

    @staticmethod
    def _comparison_features(x: pd.DataFrame) -> np.ndarray:
        dates = pd.to_datetime(x["date"], format="%Y%m%dT%H%M%S")
        days = (dates - pd.Timestamp("2015-01-01")).dt.days
        return np.column_stack([np.log(x["sqft_living"]), x["grade"], days])

    def _comparable_price(self, candidates: np.ndarray, query: np.ndarray) -> float:
        rows = self.first_sale[candidates].copy()
        for pos, home in enumerate(candidates):
            if home in self.repeats:
                sales = self.repeats[home]
                rows[pos] = sales[np.abs(self.sales[sales, 2] - query[2]).argmin()]
        differences = (self.sales[rows] - query) / self.scale
        best = np.argsort(np.square(differences).sum(axis=1), kind="stable")[:10]
        return np.nanmedian(self.sale_prices[rows[best]])

    def predict(self, x: pd.DataFrame) -> np.ndarray:
        indices = self.tree.query(
            np.deg2rad(x[["lat", "long"]].to_numpy()), k=min(max(self.n_neighbors, 50) + 1, len(self.ids)), return_distance=False
        )
        result = np.full((len(x), len(PRICE_COLUMNS)), np.nan)
        queries = self._comparison_features(x)
        for row, (own_id, candidates) in enumerate(zip(x["id"], indices, strict=True)):
            others = candidates[self.ids[candidates] != own_id]
            near = others[:self.n_neighbors]
            if len(near):
                result[row, :2] = np.nanmedian(self.prices[near], axis=0)
                if result[row, 0] > 0:
                    q25, q75 = np.nanpercentile(self.prices[near, 0], [25, 75])
                    result[row, 2] = (q75 - q25) / result[row, 0]
                result[row, 3] = self._comparable_price(others[:50], queries[row])
        return result


def _house_features(x: pd.DataFrame) -> pd.DataFrame:
    date = pd.to_datetime(x["date"], format="%Y%m%dT%H%M%S")
    angle = 2 * np.pi * (date.dt.dayofyear - 1) / (365 + date.dt.is_leap_year.astype(int))
    renovated = x["yr_renovated"].gt(0)
    bedrooms = x["bedrooms"].replace(0, np.nan)
    lot = x["sqft_lot"].replace(0, np.nan)
    return x.assign(
        sale_days=(date - pd.Timestamp("2015-01-01")).dt.days,
        sale_day_sin=np.sin(angle),
        sale_day_cos=np.cos(angle),
        age_at_sale=date.dt.year - x["yr_built"],
        years_since_renovation=date.dt.year - x["yr_renovated"].where(renovated),
        living_to_living15=x["sqft_living"] / x["sqft_living15"].replace(0, np.nan),
        lot_to_lot15=lot / x["sqft_lot15"].replace(0, np.nan),
        basement_share=x["sqft_basement"] / x["sqft_living"].replace(0, np.nan),
        living_per_bedroom=x["sqft_living"] / bedrooms,
        bathrooms_per_bedroom=x["bathrooms"] / bedrooms,
        estimated_lot_coverage=x["sqft_above"] / x["floors"].replace(0, np.nan) / lot,
    )


class HouseFeatures(TransformerMixin, BaseEstimator):
    def __init__(
        self,
        lakes: gpd.GeoDataFrame,
        mean_encode: tuple[str, ...] = ("zipcode",),
        n_neighbors: int = 20,
        n_splits: int = 3,
        random_state: int = 42,
    ) -> None:
        self.lakes = lakes
        self.mean_encode = mean_encode
        self.n_neighbors = n_neighbors
        self.n_splits = n_splits
        self.random_state = random_state

    def fit(self, X: pd.DataFrame, y: pd.Series | None = None) -> Self:
        if y is None:
            msg = "Training prices are required for neighborhood price features."
            raise ValueError(msg)
        homes = X.drop_duplicates("id")
        self.ids_ = homes["id"].to_numpy()
        self.grades_ = homes["grade"].to_numpy()
        self.years_ = homes["yr_built"].to_numpy()
        self.tree_ = BallTree(np.deg2rad(homes[["lat", "long"]].to_numpy()), metric="haversine")
        self.shoreline_ = self.lakes.to_crs(LAKE_CRS).geometry.union_all().boundary
        self.columns_ = list(self.mean_encode)
        self.price_reference_ = _NeighborPrices(X, np.asarray(y), self.n_neighbors)
        self.encoder_ = None
        if self.columns_:
            self.encoder_ = TargetEncoder(target_type="continuous", smooth="auto")
            self.encoder_.fit(X[self.columns_], y)
        return self

    def _append_location(self, x: pd.DataFrame) -> pd.DataFrame:
        coords = np.deg2rad(x[["lat", "long"]].to_numpy())
        city_km = EARTH_RADIUS_KM * haversine_distances(coords, CITY_CENTERS)
        points = gpd.GeoSeries(gpd.points_from_xy(x["long"], x["lat"]), index=x.index, crs="EPSG:4326")
        distances, candidates = self.tree_.query(coords, k=min(self.n_neighbors + 1, len(self.ids_)))
        stats = np.full((len(x), 3), np.nan)
        for i, own_id in enumerate(x["id"]):
            other = self.ids_[candidates[i]] != own_id
            near = candidates[i][other][: self.n_neighbors]
            if len(near):
                stats[i] = (
                    np.median(self.grades_[near]),
                    np.median(self.years_[near]),
                    EARTH_RADIUS_KM * distances[i][other][len(near) - 1],
                )
        return x.assign(
            distance_to_seattle_km=city_km[:, 0],
            distance_to_bellevue_km=city_km[:, 1],
            lake_distance_m=points.to_crs(LAKE_CRS).distance(self.shoreline_).to_numpy(),
            neighbor_grade_median=stats[:, 0],
            grade_vs_neighbors=x["grade"] - stats[:, 0],
            neighbor_year_median=stats[:, 1],
            year_vs_neighbors=x["yr_built"] - stats[:, 1],
            neighbor_radius_km=stats[:, 2],
        )

    def transform(self, X: pd.DataFrame) -> pd.DataFrame:
        check_is_fitted(self, "tree_")
        result = self._append_location(_house_features(X))
        result[PRICE_COLUMNS] = self.price_reference_.predict(X)
        if self.encoder_ is not None:
            result[[f"{col}_mean" for col in self.columns_]] = self.encoder_.transform(X[self.columns_])
        return result

    def fit_transform(self, X: pd.DataFrame, y: pd.Series | None = None, **fit_params: object) -> pd.DataFrame:
        self.fit(X, y, **fit_params)
        result = self._append_location(_house_features(X))
        values = X[self.columns_]
        target = np.asarray(y)
        encoded = np.empty(values.shape)
        prices = np.empty((len(X), len(PRICE_COLUMNS)))
        folds = GroupKFold(n_splits=self.n_splits, shuffle=True, random_state=self.random_state)
        for train, valid in folds.split(X, groups=X["id"]):
            reference = _NeighborPrices(X.iloc[train], target[train], self.n_neighbors)
            prices[valid] = reference.predict(X.iloc[valid])
            if self.columns_:
                encoder = TargetEncoder(target_type="continuous", smooth="auto")
                encoder.fit(values.iloc[train], target[train])
                encoded[valid] = encoder.transform(values.iloc[valid])
        result[PRICE_COLUMNS] = prices
        if self.columns_:
            result[[f"{col}_mean" for col in self.columns_]] = encoded
        return result
```

```python
class _AnchorRidges:
    def __init__(self, frame: pd.DataFrame, y: np.ndarray, columns: tuple[str, ...], k: int) -> None:
        self.columns = list(columns)
        if not self.columns:
            msg = "No features selected from price_corr."
            raise ValueError(msg)
        self.medians = frame[self.columns].median()
        inputs = self._inputs(frame)
        targets = y[:, None] / frame[["sqft_living", "sqft_lot"]].to_numpy()
        anchors = frame.drop_duplicates("id")
        self.ids = anchors["id"].to_numpy()
        coords = np.deg2rad(anchors[["lat", "long"]].to_numpy())
        self.tree = BallTree(coords, metric="haversine")
        candidates = self.tree.query(coords, k=min(k + 1, len(anchors)), return_distance=False)
        sales = frame.groupby("id").indices
        self.models = []
        self.training_ids = []
        anchors = zip(self.ids, candidates, strict=True)
        for own_id, indices in tqdm(anchors, total=len(self.ids), desc="Локальные Ridge", unit="дом", leave=False):
            neighbors = self.ids[indices]
            neighbors = neighbors[neighbors != own_id][:k]
            rows = np.concatenate([sales[neighbor] for neighbor in neighbors])
            model = make_pipeline(StandardScaler(), Ridge(alpha=1.0))
            self.models.append(model.fit(inputs[rows], targets[rows]))
            self.training_ids.append(set(neighbors))

    def _inputs(self, frame: pd.DataFrame) -> np.ndarray:
        return frame[self.columns].fillna(self.medians).to_numpy()

    def predict(self, frame: pd.DataFrame) -> np.ndarray:
        coords = np.deg2rad(frame[["lat", "long"]].to_numpy())
        nearest = self.tree.query(coords, k=1, return_distance=False).ravel()
        inputs = self._inputs(frame)
        predictions = np.empty((len(frame), 2))
        for i, (own_id, nearest_anchor) in enumerate(zip(frame["id"], nearest, strict=True)):
            anchor = nearest_anchor
            if own_id in self.training_ids[anchor]:
                ordered = self.tree.query(coords[i : i + 1], k=len(self.ids), return_distance=False)[0]
                anchor = next(j for j in ordered if own_id not in self.training_ids[j])
            predictions[i] = self.models[anchor].predict(inputs[i : i + 1])[0]
        return predictions


RIDGE_COLUMNS = ["ridge_price_per_sqft_living", "ridge_price_per_sqft_lot"]


class LocalRidgeFeatures(TransformerMixin, BaseEstimator):
    def __init__(self, house_features: TransformerMixin, columns: tuple[str, ...]) -> None:
        self.house_features = house_features
        self.columns = columns

    def _fit_reference(self, x: pd.DataFrame, y: np.ndarray) -> pd.DataFrame:
        self.features_ = clone(self.house_features)
        frame = self.features_.fit_transform(x, y)
        self.anchors_ = _AnchorRidges(frame, y, self.columns, self.features_.n_neighbors)
        self.selected_features_ = self.anchors_.columns
        return frame

    def fit(self, X: pd.DataFrame, y: pd.Series) -> Self:
        self._fit_reference(X, np.asarray(y))
        return self

    def transform(self, X: pd.DataFrame) -> pd.DataFrame:
        check_is_fitted(self, "anchors_")
        frame = self.features_.transform(X)
        frame[RIDGE_COLUMNS] = self.anchors_.predict(frame)
        return frame

    def fit_transform(self, X: pd.DataFrame, y: pd.Series | None = None, **fit_params: object) -> pd.DataFrame:
        if y is None:
            msg = "Training prices are required for local Ridge features."
            raise ValueError(msg)
        target = np.asarray(y)
        result = self._fit_reference(X, target)
        folds = GroupKFold(n_splits=self.features_.n_splits, shuffle=True, random_state=self.features_.random_state)
        predictions = np.empty((len(X), len(RIDGE_COLUMNS)))
        splits = folds.split(X, groups=X["id"])
        for train, valid in tqdm(splits, total=folds.get_n_splits(), desc="OOF Ridge", unit="фолд"):
            local = clone(self).fit(X.iloc[train], target[train], **fit_params)
            predictions[valid] = local.transform(X.iloc[valid])[RIDGE_COLUMNS].to_numpy()
        result[RIDGE_COLUMNS] = predictions
        return result
```
