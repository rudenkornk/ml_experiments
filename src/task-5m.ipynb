Рассмотрим линейную регрессию с одним признаком $y(x) = \theta_0 + \theta_1 x$. Пусть несмещенный, нормальный, гомоскедастичный. Получите формулы доверительных интервалов для параметров $\theta_0$ и $\theta_1$.


<br/>

<details style="box-sizing:border-box;max-width:100%;margin:0 0 18px;padding:8px 14px 10px;border:1px solid rgba(199,50,78,0.20);border-radius:10px;color:inherit;">

<summary style="cursor:pointer;color:inherit;font-size:13px;font-weight:750;text-decoration:underline;text-decoration-color:#C7324E;text-decoration-thickness:1.5px;text-underline-offset:3px;">Кликни для показа подсказки</summary>

Достаточно воспользоваться формулами с лекции и расписать в них произведение матриц для модели с одним признаком.

</details>

$$
y = X\theta + \varepsilon,
\qquad
\varepsilon \sim \mathcal{N}(0,\sigma^2 I_n).
$$

$$
\begin{aligned}
A &= (X^\top X)^{-1},\\
\hat{\theta} &= AX^\top y,\\
e &= y-X\hat{\theta},\\
\hat{\sigma}^2 &= \frac{\|e\|^2}{n-d},\\
t_* &= t_{n-d,\,\frac{1+\alpha}{2}}.
\end{aligned}
$$

Доверительный интервал для $\theta_j$ уровня $\alpha$:

$$
\theta_j \in
\left[
\hat{\theta}_j-t_*\hat{\sigma}\sqrt{A_{jj}},
\;
\hat{\theta}_j+t_*\hat{\sigma}\sqrt{A_{jj}}
\right].
$$

---

Для линейной регрессии с одним признаком:

$$
y = \theta_0\mathbf{1}+x\theta_1+\varepsilon,
\qquad
x =
\begin{pmatrix}
x_1\\
\vdots\\
x_n
\end{pmatrix},
\qquad
X = \begin{pmatrix}\mathbf{1} & x\end{pmatrix}.
$$

$$
X^\top X
=
\begin{pmatrix}
\mathbf{1}^\top\\
x^\top
\end{pmatrix}
\begin{pmatrix}
\mathbf{1} & x
\end{pmatrix}
=
n
\begin{pmatrix}
1 & \bar{x}\\
\bar{x} & \dfrac{x^\top x}{n}
\end{pmatrix}.
$$

$$
S = \sum_{i=1}^{n}(x_i-\bar{x})^2
  = x^\top x-n\bar{x}^2.
$$

$$
\begin{aligned}
A
&= (X^\top X)^{-1}\\
&=
\frac{1}{x^\top x-n\bar{x}^2}
\begin{pmatrix}
\dfrac{x^\top x}{n} & -\bar{x}\\
-\bar{x} & 1
\end{pmatrix}\\
&=
\frac{1}{S}
\begin{pmatrix}
\dfrac{x^\top x}{n} & -\bar{x}\\
-\bar{x} & 1
\end{pmatrix}.
\end{aligned}
$$

$$
\begin{aligned}
\hat{\theta}
&= AX^\top y\\
&=
\frac{1}{S}
\begin{pmatrix}
\dfrac{x^\top x}{n} & -\bar{x}\\
-\bar{x} & 1
\end{pmatrix}
\begin{pmatrix}
\mathbf{1}^\top\\
x^\top
\end{pmatrix}y\\
&=
\frac{1}{S}
\begin{pmatrix}
\dfrac{x^\top x}{n} & -\bar{x}\\
-\bar{x} & 1
\end{pmatrix}
\begin{pmatrix}
n\bar{y}\\
x^\top y
\end{pmatrix}\\
&=
\frac{1}{S}
\begin{pmatrix}
x^\top x\,\bar{y}-\bar{x}\,x^\top y\\
-n\bar{x}\bar{y}+x^\top y
\end{pmatrix}.
\end{aligned}
$$

Доверительные интервалы уровня $\alpha$
(знак $\pm$ применяется отдельно к каждой координате):

$$
\begin{pmatrix}
\theta_0\\
\theta_1
\end{pmatrix}
\in
\frac{1}{S}
\begin{pmatrix}
x^\top x\,\bar{y}-\bar{x}\,x^\top y\\
-n\bar{x}\bar{y}+x^\top y
\end{pmatrix}
\;\pm\;
t_{n-2,\,\frac{1+\alpha}{2}}\hat{\sigma}
\begin{pmatrix}
\sqrt{\dfrac{x^\top x}{nS}}\\
\dfrac{1}{\sqrt{S}}
\end{pmatrix}.
$$

$$
\hat{\sigma}^2
=
\frac{\|y-X\hat{\theta}\|^2}{n-2}.
$$

Пусть $X_1,...,X_n$ &mdash; выборка (независимые одинаково распределенные случайные величины) из экспоненциального распределения с плотностью $p_\theta(x) = \theta e^{-\theta x},$ где $x>0, \theta>0$.

> Такое распределение описывает *время до наступления некоторого события* в условиях, когда события происходят независимо друг от друга. Например, таким событием может быть поступление очередного заказа таксисту или курьеру службы доставки. В таком случае $X_i$ &mdash; время ожидания $i$-го заказа после завершения предыдущего.
> 
> Параметр $\theta$ имеет смысл *интенсивности заказов в единицу времени*. Следовательно, чем больше $\theta$, тем меньше среднее значение $X_i$, а точнее $\mathsf{E}_\theta X_i = 1/\theta$. Иначе говоря, если приходит в среднем $\theta=15$ заказов в час, то среднее время между заказами составляет $\mathsf{E}_\theta X_i = 1/\theta = 1/15$ часа или $4$ минуты.
>
> Интересный факт: экспоненциальное распределение является единственным непрерывным распределением, обладающим *свойством отсутствия памяти*: для любых $s,t > 0$ выполнено $\mathsf{P}_{\theta}(X_1 > t + s \mid X_1 > s) = \mathsf{P}_{\theta}(X_1 > t)$. Свойство говорит о том, что если следующего заказа ожидаем уже $s$ минут после
получения предыдущего, то в среднем ждать осталось столько же, как будто бы предыдущий заказ только что пришел.
> 
> При выполнении задач используйте следующий факт: $\mathsf{E}_\theta X_i^k = k!/\theta^k$.

**1.** Постройте график плотности экспоненциального распределения.

```python
import numpy as np
import matplotlib.pyplot as plt

rate = 1.0

x = np.linspace(0, 6 / rate, 1000)
density = rate * np.exp(-rate * x)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, density, label=f"Интенсивность λ = {rate:g}")
ax.set(
    xlabel="x",
    ylabel="Плотность",
    title="Экспоненциальное распределение",
    xlim=(0, x[-1]),
    ylim=(0, None),
)
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
plt.show()
```
