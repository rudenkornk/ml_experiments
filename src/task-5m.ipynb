\varepsilon \sim \mathcal{N}(0,\sigma^2 I_n).

$$
\begin{aligned}
A &= (X^\top X)^{-1},\\
\hat{\theta} &= AX^\top y,\\

\hat{\sigma}^2 &= \frac{\|e\|^2}{n-d},\\
t_* &= t_{n-d,\,\frac{1+\alpha}{2}}.

Доверительный интервал для $\theta_j$ уровня $\alpha$:

$$
\theta_j \in

Для линейной регрессии с одним признаком:

$$
y = \theta_0\mathbf{1}+x\theta_1+\varepsilon,
\qquad
x =
\begin{pmatrix}
x_1\\
\vdots\\
x_n
\end{pmatrix},
\qquad
X = \begin{pmatrix}\mathbf{1} & x\end{pmatrix}.
$$

$$
X^\top X
=
\begin{pmatrix}
\mathbf{1}^\top\\
x^\top
\end{pmatrix}
\begin{pmatrix}
\mathbf{1} & x
\end{pmatrix}
=
n
\begin{pmatrix}
1 & \bar{x}\\
\bar{x} & \dfrac{x^\top x}{n}
\end{pmatrix}.
$$

$$
S = \sum_{i=1}^{n}(x_i-\bar{x})^2
  = x^\top x-n\bar{x}^2.
$$

$$
\begin{aligned}
A
&= (X^\top X)^{-1}\\
&=
\frac{1}{x^\top x-n\bar{x}^2}
\begin{pmatrix}
\dfrac{x^\top x}{n} & -\bar{x}\\
-\bar{x} & 1
\end{pmatrix}\\

\frac{1}{S}
\begin{pmatrix}
\dfrac{x^\top x}{n} & -\bar{x}\\
-\bar{x} & 1
\end{pmatrix}.
\end{aligned}
$$

$$
\begin{aligned}
\hat{\theta}
&= AX^\top y\\
&=
\frac{1}{S}
\begin{pmatrix}
\dfrac{x^\top x}{n} & -\bar{x}\\
-\bar{x} & 1
\end{pmatrix}
\begin{pmatrix}
\mathbf{1}^\top\\
x^\top
\end{pmatrix}y\\

\end{pmatrix}

x^\top y
\end{pmatrix}\\

\frac{1}{S}
\begin{pmatrix}
x^\top x\,\bar{y}-\bar{x}\,x^\top y\\

Пусть $X_1,...,X_n$ &mdash; выборка (независимые одинаково распределенные случайные величины) из экспоненциального распределения с плотностью $p_\theta(x) = \theta e^{-\theta x},$ где $x>0, \theta>0$.

> Такое распределение описывает *время до наступления некоторого события* в условиях, когда события происходят независимо друг от друга. Например, таким событием может быть поступление очередного заказа таксисту или курьеру службы доставки. В таком случае $X_i$ &mdash; время ожидания $i$-го заказа после завершения предыдущего.
> 
> Параметр $\theta$ имеет смысл *интенсивности заказов в единицу времени*. Следовательно, чем больше $\theta$, тем меньше среднее значение $X_i$, а точнее $\mathsf{E}_\theta X_i = 1/\theta$. Иначе говоря, если приходит в среднем $\theta=15$ заказов в час, то среднее время между заказами составляет $\mathsf{E}_\theta X_i = 1/\theta = 1/15$ часа или $4$ минуты.
>
> Интересный факт: экспоненциальное распределение является единственным непрерывным распределением, обладающим *свойством отсутствия памяти*: для любых $s,t > 0$ выполнено $\mathsf{P}_{\theta}(X_1 > t + s \mid X_1 > s) = \mathsf{P}_{\theta}(X_1 > t)$. Свойство говорит о том, что если следующего заказа ожидаем уже $s$ минут после
получения предыдущего, то в среднем ждать осталось столько же, как будто бы предыдущий заказ только что пришел.
> 
> При выполнении задач используйте следующий факт: $\mathsf{E}_\theta X_i^k = k!/\theta^k$.

**1.** Постройте график плотности экспоненциального распределения.

```python
import numpy as np
import matplotlib.pyplot as plt

rate = 1.0

x = np.linspace(0, 6 / rate, 1000)
density = rate * np.exp(-rate * x)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, density, label=f"Интенсивность λ = {rate:g}")
ax.set(
    xlabel="x",
    ylabel="Плотность",
    title="Экспоненциальное распределение",
    xlim=(0, x[-1]),
    ylim=(0, None),
)
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
plt.show()
```

### Дельта-метод

Пусть $T_n$ — оценка $h^{-1}(\theta)$ и
$$
\sqrt{n}\left(T_n-h^{-1}(\theta)\right)
\xrightarrow{d}
\mathcal{N}(0,\sigma^2).
$$

По формуле Тейлора:
$$
h(T_n)-h\left(h^{-1}(\theta)\right)
\approx
h'\left(h^{-1}(\theta)\right)
\left(T_n-h^{-1}(\theta)\right).
$$

Следовательно, при $h'(h^{-1}(\theta))\ne 0$:
$$
\sqrt{n}\,
\frac{h(T_n)-h\left(h^{-1}(\theta)\right)}
{h'\left(h^{-1}(\theta)\right)}
\xrightarrow{d}
\mathcal{N}(0,\sigma^2).
$$

Поскольку $h(h^{-1}(\theta))=\theta$, получаем:
$$
\sqrt{n}\left(h(T_n)-\theta\right)
\xrightarrow{d}
\mathcal{N}\left(
0,\,
\sigma^2\left[h'\left(h^{-1}(\theta)\right)\right]^2
\right).
$$

$$
F_{X^2}(t)

Вычислим моменты:
$$
\mathbb{E}[X^2]

=\frac{24}{\theta^4}.
$$

Следовательно,
$$
\operatorname{Var}(X^2)
=\mathbb{E}[X^4]-\left(\mathbb{E}[X^2]\right)^2
=\frac{24}{\theta^4}-\frac{4}{\theta^4}

### Применение дельта-метода

### Гамма-распределение

Плотность гамма-распределения с параметром формы $a$ и масштаба $\sigma$:
$$
p(x)=\frac{1}{\sigma^a\Gamma(a)}x^{a-1}e^{-x/\sigma},
\qquad x>0.
$$

Для $T_n=\overline{X}$:
$$
a=n,
\qquad
\sigma=\frac{1}{n\theta}.
$$

Следовательно,
$$
p_{\overline{X}}(x)
=\frac{n^n\theta^n}{\Gamma(n)}x^{n-1}e^{-xn\theta}
=\frac{n^n}{\Gamma(n)x}(x\theta)^n e^{-n(x\theta)},
\qquad x>0.
$$

---

### Плотность при умножении на константу

При $c>0$:
$$
F_X(t)=\mathbb{P}(X<t),
\qquad
p_X(t)=F_X'(t).
$$

$$
F_{cX}(t)
=\mathbb{P}\left(X<\frac{t}{c}\right)
=F_X\left(\frac{t}{c}\right).
$$

Поэтому
$$
p_{cX}(t)=\frac{1}{c}\,p_X\left(\frac{t}{c}\right).
$$

---

### Распределение $Y=\theta\overline{X}$

$$
p_Y(x)
=\frac{1}{\theta}\,
p_{\overline{X}}\left(\frac{x}{\theta}\right)
=\frac{n^n}{\Gamma(n)}x^{n-1}e^{-nx},
\qquad x>0.
$$

Таким образом,
$$
Y=\theta\overline{X}
\sim \operatorname{Gamma}\left(
a=n,\ \mathrm{scale}=\frac{1}{n}
\right).
$$

**7.** Мы построили три доверительных интервала для параметра $\theta$ (пункты 3, 5, 6). Какой из них лучше? 

Повторите эксперимент, показанный на занятии, 
* построив визуализации доверительны интервалов при увеличении размера выборки,
* оценив долю покрытия параметра интервалом.

Кроме этого, оцените и визуализируйте также среднюю длину каждого интервала в зависимости от размера выборки.

 Точная оценка (по гамма распределению) работает при любом n. Интервалы Вальда таким свойством не обладают.  
 У первого способа с интервалом Вальда более узкий интервал при том же доверительном диапазоне, он немного точнее.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as sps

from confidence_intervals_tools import draw_confidence_interval


theta = 1.0
alpha = 0.95
num_experiments = 10_000
max_n = 500
seed = 42

methods = (
    "Вальд по среднему",
    "Вальд по среднему квадратов",
    "Точный гамма-интервал",
)
colors = ("tab:blue", "tab:orange", "tab:green")
z = sps.norm.ppf((1 + alpha) / 2)


def calculate_intervals(mean_x, mean_x2, n):
    theta_mean = 1 / mean_x
    theta_squares = np.sqrt(2 / mean_x2)

    delta_mean = z * theta_mean / np.sqrt(n)
    delta_squares = z * np.sqrt(5 / 4) * theta_squares / np.sqrt(n)

    q_left = sps.gamma.ppf((1 - alpha) / 2, a=n, scale=1 / n)
    q_right = sps.gamma.ppf((1 + alpha) / 2, a=n, scale=1 / n)

    return {
        methods[0]: (theta_mean - delta_mean, theta_mean + delta_mean),
        methods[1]: (
            theta_squares - delta_squares,
            theta_squares + delta_squares,
        ),
        methods[2]: (q_left / mean_x, q_right / mean_x),
    }


rng_path = np.random.default_rng(seed)
sample = rng_path.exponential(scale=1 / theta, size=max_n)

n_values = np.arange(1, max_n + 1)
mean_x = np.cumsum(sample) / n_values
mean_x2 = np.cumsum(sample**2) / n_values

intervals = calculate_intervals(mean_x, mean_x2, n_values)
estimates = {
    methods[0]: 1 / mean_x,
    methods[1]: np.sqrt(2 / mean_x2),
    methods[2]: None,
}
```

```python
fig = plt.figure(figsize=(12, 11))

for index, (method, color) in enumerate(zip(methods, colors, strict=True), 1):
    ax = fig.add_subplot(3, 1, index)
    left, right = intervals[method]

    draw_confidence_interval(
        left=left,
        right=right,
        estimation=estimates[method],
        color_estimation=color,
        color_interval=color,
        label_estimation="Оценка θ",
        label_interval=f"Интервал уровня {alpha:.0%}",
    )
    ax.axhline(theta, color="black", linestyle="--", label="Истинное θ")
    ax.set(
        title=method,
        xlabel="Размер выборки n",
        ylabel="Значение параметра",
        xlim=(1, max_n),
    )
    ax.grid(alpha=0.3)
    ax.legend()

fig.tight_layout()
plt.show()
```

```python
sample_sizes = np.unique(
    np.concatenate([
        np.arange(2, 31),
        np.geomspace(32, max_n, 35).astype(int),
    ])
)

rng_simulation = np.random.default_rng(seed + 1)
records = []

for n in sample_sizes:
    samples = rng_simulation.exponential(
        scale=1 / theta,
        size=(num_experiments, n),
    )
    mean_x = samples.mean(axis=1)
    mean_x2 = np.square(samples).mean(axis=1)

    for method, (left, right) in calculate_intervals(mean_x, mean_x2, n).items():
        covered = (left <= theta) & (theta <= right)

        records.append({
            "n": n,
            "method": method,
            "coverage": covered.mean(),
            "mean_length": (right - left).mean(),
        })

results = pd.DataFrame(records)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for method, color in zip(methods, colors, strict=True):
    values = results.loc[results["method"] == method]

    axes[0].plot(
        values["n"],
        values["coverage"],
        color=color,
        label=method,
    )
    axes[1].plot(
        values["n"],
        values["mean_length"],
        color=color,
        label=method,
    )

axes[0].axhline(
    alpha,
    color="black",
    linestyle="--",
    label=f"Заданный уровень {alpha:.0%}",
)
axes[0].set(
    title="Доля покрытия параметра",
    xlabel="Размер выборки n",
    ylabel="Доля покрытия",
    ylim=(0.8, 1),
)
axes[1].set(
    title="Средняя длина интервала",
    xlabel="Размер выборки n",
    ylabel="Средняя длина",
)

for ax in axes:
    ax.set_xscale("log")
    ax.grid(alpha=0.3)
    ax.legend()

fig.tight_layout()
plt.show()
```
