### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 1</span>

Рассмотрим модель логистической регрессии. Признаки объекта представимы в виде $d$-мерного вектора $x \in \mathbb{R}^d$, класс имеет бернулиевское распределение $Y \sim Bern(\mu_\theta(x))$. Мы делаем следующее предположение о зависимости параметра вероятности от признаков:
$${\mu_\theta(x) = \sigma(x^T\theta)= \cfrac{1}{1 + e^{-x^T\theta}}}.$$

Зададим регуляризацию через минимизацию следующего функционала:

$$F(\theta) = -\frac{1}{n}\sum_{i=1}^n \left[Y_i \log{\sigma(\theta^T x_i)} + (1 - Y_i) \log{\left(1 - \sigma(\theta^T x_i)\right)}\right] + \lambda\theta^T \theta.$$

**1.** Выведите формулы градиентного спуска (GD) и стохастического градиентного спуска (SGD).

Функция потерь логистической регрессии с $L_2$-регуляризацией:

$$
F(\theta)
=
-\frac{1}{n}\sum_{i=1}^{n}
\left[
y_i\log\sigma(\theta^\top x_i)
+
(1-y_i)\log\left(1-\sigma(\theta^\top x_i)\right)
\right]
+
\lambda\theta^\top\theta.
$$

Обозначим первую часть функции потерь через $F_1(\theta)$.
Из лекции по логистической регрессии:

$$
\nabla F_1(\theta)
=
\frac{1}{n}X^\top\left(\sigma(X\theta)-Y\right).
$$

Следовательно,

$$
\nabla F(\theta)
=
\frac{1}{n}X^\top\left(\sigma(X\theta)-Y\right)
+
2\lambda\theta.
$$

**Градиентный спуск (GD):**

$$
\theta^{(t+1)}
=
\theta^{(t)}
-
\eta\left[
\frac{1}{n}X^\top
\left(\sigma(X\theta^{(t)})-Y\right)
+
2\lambda\theta^{(t)}
\right].
$$

**Стохастический градиентный спуск (SGD) с мини-батчем:**

Пусть $\mathcal I$ — множество индексов объектов мини-батча,
$k=|\mathcal I|$.

$$
\theta^{(t+1)}
=
\theta^{(t)}
-
\eta\left[
\frac{1}{k}X_{\mathcal I}^\top
\left(\sigma(X_{\mathcal I}\theta^{(t)})-Y_{\mathcal I}\right)
+
2\lambda\theta^{(t)}
\right].
$$

**2.** Покажите, что $F(\theta)$ &mdash; выпуклая функция по $\theta$ и, как следствие, имеет единственный экстремум, являющийся глобальным максимумом. *Указание*. Посчитайте гессиан (матрицу вторых производных) и покажите, что она положительно определена.


Также из лекции:

$$
\nabla^2 F_1(\theta)
=
\frac{1}{n}X^\top W X,
$$

где

$$
W
=
\operatorname{diag}\left(\sigma(X\theta)\right)
\operatorname{diag}\left(1-\sigma(X\theta)\right).
$$

Таким образом,

$$
\nabla^2 F(\theta)
=
\frac{1}{n}X^\top W X + 2\lambda I.
$$

Матрица $X^\top W X$ симметрична и положительно полуопределена.
При $\lambda>0$ матрица $2\lambda I$ положительно определена,
поэтому гессиан $\nabla^2 F(\theta)$ положительно определён, притом во всех точках, т.е. функция выпуклая. Соответственно у нее есть единственный экстремум.

**3.** Опишите, как может вести себя решение при отсутствии регуляризации, то есть при $\lambda = 0$

При $\lambda=0$ гессиан становится лишь неотрицательно определенным, и теорема о достаточном условии локального экстремума не выполняется.
Функция не обязана быть строго выпукла во всех точках и может иметь неоднозначные решения.
