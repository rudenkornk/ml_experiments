**Процедура предобработки данных:**

**a).** Зациклим season, mnth, weekday  
**b).** holiday, workingday, weathersit -> onehot.
**c).** temp уберем, оставим только atemp.

Реализуйте данную процедуру.

```python
cyclic_columns = {
    "season": (4, 1),
    "mnth": (12, 1),
    "weekday": (7, 0),
}

categorical_columns = ["holiday", "workingday", "weathersit"]

# Фиксируем категории по train, чтобы кодирование train и test совпадало.
categories = {
    column: sorted(train[column].unique())
    for column in categorical_columns
}

def preprocess(data):
    processed = data.copy()

    for column, (period, minimum) in cyclic_columns.items():
        value = (processed[column] - minimum) % period
        processed[f"{column}_sin"] = np.sin(2 * np.pi * value / period)
        processed[f"{column}_cos"] = np.cos(2 * np.pi * value / period)

    processed = processed.drop(columns=[*cyclic_columns, "temp"])

    for column in categorical_columns:
        processed[column] = pd.Categorical(
            processed[column],
            categories=categories[column],
        )

    return pd.get_dummies(
        processed,
        columns=categorical_columns,
        drop_first=True,
        dtype=int,
    )

train_processed = preprocess(train)
test_processed = preprocess(test)

# На случай, если какая-то категория не встретилась в test.
test_processed = test_processed.reindex(columns=train_processed.columns, fill_value=0)
```
