* `slope` — наклон пика упражнения сегмента ST (0: восходящий, 1: плоский, 2: нисходящий);
* `ca` — количество основных сосудов (0-3), окрашенных флуороскопией;
* `thal` — талассемия (3 = норма; 6 = исправленный дефект; 7 = обратимый дефект);
* `target` — наличие сердечного заболевания (0 = нет; 1-4: различные стадии заболевания).

**1.** Разделите данные на обучающую и тестовую части в соотношении 80:20, используйте параметры `stratify` и `random_state`.

```python
path = kagglehub.dataset_download("redwankarimsony/heart-disease-data")
```

```python
print(path)
```

```python
df = pd.read_csv(f"{path}/heart_disease_uci.csv")
df.head()
```

```python
# Dataset is biased by sex, so lets stratify by it too.
train, test = train_test_split(
    df,
    test_size=0.2,
    stratify=df[["num", "sex"]],
    random_state=0,
)
train_x = train.drop(columns="num")
train_y = train["num"]
test_x = test.drop(columns="num")
test_y = test["num"]
```

**2.** Проведите небольшой разведочный анализ данных (EDA). Примеры можно посмотреть в ноутбуке с первого занятия. Слишком подробный анализ не требуется, достаточно базового, но качественно оформленного с красивыми и информативными графиками, а также с выводами по ним.

На основе анализа выполните необходимые преобразования признаков.

```python
train.info()
display(train.describe(include='all'))
```

```python
features = ["age", "trestbps", "chol", "thalch", "oldpeak"]
cols = len(features)

fig, axes = plt.subplots(1, cols, figsize=(15, 4), squeeze=False)

for feature, ax in zip(features, axes.flat):
    sns.histplot(train[feature], kde=True, ax=ax)
    ax.set_title(feature)

for ax in axes.flat[len(features):]:
    ax.remove()

plt.tight_layout()
plt.show()
```

Все характеристики имеют околонормальное распределение. За исключением chol и oldpeak, которые имеют пики в 0. Для холестерина это скорее всего означает отсутствие данных. Для oldpeak 0 это нормальное типичное значение.

```python
features = ["sex", "dataset", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal", "num"]
cols = len(features) // 2
rows = 2

fig, axes = plt.subplots(rows, cols, figsize=(15, 4 * rows), squeeze=False)

for feature, ax in zip(features, axes.flat):
    sns.countplot(data=train, x=feature, ax=ax, palette="pastel", hue=feature, legend=False)
    ax.set_title(feature)
    ax.tick_params(axis="x", rotation=45)
    
plt.tight_layout()
plt.show()
```

```python
sns.clustermap(train.corr(numeric_only=True), cmap="coolwarm", annot=False, figsize=(12, 10))
plt.show()
```

```python
categorical = ["sex", "dataset", "cp", "restecg", "slope", "thal"]
numeric = ["age", "trestbps", "chol", "thalch", "oldpeak", "fbs", "exang", "ca"]

preprocessor = ColumnTransformer(
    [
        ("numeric", make_pipeline(SimpleImputer(strategy="median"), MinMaxScaler()), numeric),
        (
            "categorical",
            make_pipeline(
                SimpleImputer(strategy="most_frequent"), OneHotEncoder(handle_unknown="ignore", sparse_output=False)
            ),
            categorical,
        ),
    ]
)
preprocessor.set_output(transform="pandas")

encoder = OneHotEncoder(handle_unknown="ignore")

train_x_preprocessed = preprocessor.fit_transform(train_x)
test_x_preprocessed = preprocessor.transform(test_x)
```

```python
train_x_preprocessed
```

Выводы: в целом хорошие достаточно полные денные. С таргетом положительно коррелируют количество окрашенных основных сосудов, возраст и депрессия ST.
Отрицательно коррелирует холестерин и максимальная достигнутая частота сердечных сокращений.

**3.** Обучите тривиальную модель и оцените ее качество по тестовой выборке.

```python
const_model = DummyClassifier(strategy="most_frequent")
const_model.fit(train_x_preprocessed, train_y)
print(accuracy_score(train_y, const_model.predict(train_x_preprocessed)))
print(accuracy_score(test_y, const_model.predict(test_x_preprocessed)))
```

**4.** Обучите модель kNN для ряда различных значений гиперпараметра $k$ — количество соседей. Сравните качество модели для разных $k$ на обучающей и тестовой частях данных.

```python
k = range(1,100)
train_result = []
test_result = []
for kinst in k:
    model = KNeighborsClassifier(n_neighbors=kinst, algorithm="brute")
    model.fit(train_x_preprocessed, train_y)
    train_result.append(accuracy_score(train_y, model.predict(train_x_preprocessed)))
    test_result.append(accuracy_score(test_y, model.predict(test_x_preprocessed)))
```

```python
plt.scatter(k, train_result, label="train score")
plt.scatter(k, test_result, label="test score")
plt.xlabel("k")
plt.ylabel("Score")
plt.show()
print(test_result[30])
```

Сравните kNN с оптимальным значением гиперпараметра $k$ с тривиальной моделью из предыдущего пункта.

Dummy выдает около 0.45, kNN с оптимальным k ~= 0.6

**Выводы:**

1. Для kNN не забывать кодировать категориальные данные.
2. Каким-то образом учитывать NaN, например most_frequent или median.
3. Не забывать нормировать данные.
4. kNN работает лучше чем константа.
5. Оптимальное значение k в задаче ~ 20-50
