### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 1</span>

Докажите, что метод KMeans для евклидовой метрики делит все пространство объектов на выпуклые многоугольники, возможно, неограниченные.

Если вы используете какие-то теоремы и свойства, нужно их четко сформулировать, определить все обозначения, а также привести четкое обоснование, как вы пользуетесь соответствующими свойствами.

$K$-means: $X_1,\ldots,X_n$ — датасет; $X_i$ — вектор признаков.

$S=\{S_1,\ldots,S_K\}$ — разбиение на кластеры.

Задача:

$$
\underset{S}{\arg\min}
\sum_{i=1}^{K}\sum_{X\in S_i}\lVert X-\mu_i\rVert^2,
$$

где

$$
\mu_i=\frac{1}{|S_i|}\sum_{X\in S_i}X.
$$

При этом кластер для точки $X$

$$
k=\underset{i}{\arg\min}\,\lVert X-\mu_i\rVert^2.
$$

Докажем требуемое а-ля по индукции, рассматривая конкретный кластер (без потери общности) $S_1$ с центром $\mu_1$.

1. Во-первых, заметим, что любые два центра делят всю плоскость на две части:

   $$
   \begin{array}{c|c}
   \qquad S_1 \qquad & \qquad S_2 \qquad \\[8pt]
   \mu_1\;\bullet\,\cdots\,\cdots &
   \cdots\,\cdots\,\bullet\;\mu_2 \\[8pt]
   \vphantom{S_1} & \vphantom{S_2}
   \end{array}
   $$

2. База: рассмотрим $\mu_1$ без всех остальных центров. Он «делит» плоскость на бесконечный $0$-угольник. Добавляя $\mu_2$, новый центр «отрезает» $1$ сторону бесконечной прямой, и у $S_1$ остаётся выпуклый $1$-угольник.

3. Индукция: пусть на каком-то шаге $n-1$ к $S_1$ принадлежит $K_{n-1}$ — выпуклый многоугольник. Добавим центр $\mu_n$. Тогда он «отберёт» путём «отрезания» с помощью прямой у $\mu_1$ какую-то часть. Но у выпуклого многоугольника прямая может отрезать часть, только получив после этого снова выпуклый многоугольник. (Надеюсь, это можно принять без доказательства...)

4. $\Rightarrow$ На всех шагах у $S_1$ будет выпуклая многоугольная область пространства.
