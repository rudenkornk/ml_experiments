### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 1</span>

Докажите, что индекс Джини равен вероятности ошибки случайного классификатора, который выдаёт предсказания с вероятностями пропорционально доле классов в выборке.

Доли классов: $p_1,\ldots,p_K$, где
$$
\sum_{i=1}^{K}p_i=1.
$$

Непересекающиеся случаи предсказания классов $i$ дают вероятность ошибки

$$
P_{\mathrm{err}}
=
\sum_{i=1}^{K}p_i\,P(X\ne i).
$$

Поскольку
$$
P(X\ne i)=1-P(X=i)=1-p_i,
$$

получаем
$$
P_{\mathrm{err}}
=
\sum_{i=1}^{K}p_i(1-p_i)
=
1-\sum_{i=1}^{K}p_i^2.
$$

Это **индекс Джини**.

### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 2</span>

Пусть $X = (x_1, ..., x_n)$ &mdash; выборка объектов и $Y = (Y_1, ..., Y_n)$ &mdash; соответствующие значения вещественного таргета. Мера неоднородности для набора объектов вычисляется на основе того, насколько хорошо их таргет предсказывается константой:

$$
H(X) = \min_{c \in \mathscr{Y}} \dfrac{1}{n} \sum_{i=1}^n L(Y_i, c),
$$

где $L(y, c)$ &mdash; некоторая функция потерь, $\mathscr{Y}$ &mdash; множество значений таргета. Соответственно, чтобы получить аналитическое выражение для меры однородности при конкретной функции потерь, необходимо аналитически найти оптимальное значение константы и подставить его в формулу для $H(X)$.

Выведите аналитическое выражение для меры неоднородности и оптимальное значение в листьях дерева для следующих функций потерь:

**1.** $L(y, c) = (y-c)^2$

$$
X=(x_1,\ldots,x_n),
\qquad
Y=(Y_1,\ldots,Y_n).
$$

$$
H(X)
=
\min_{c\in\mathcal{Y}}
\frac{1}{n}\sum_{i=1}^{n}L(Y_i,c).
$$

### 1. Квадратичная функция потерь

$$
L(y,c)=(y-c)^2.
$$

$$
H(X)
=
\min_{c\in\mathbb{R}}
\frac{1}{n}\sum_{i=1}^{n}(Y_i-c)^2.
$$

Условие минимума:
$$
\frac{d}{dc}
\left(
\frac{1}{n}\sum_{i=1}^{n}(Y_i-c)^2
\right)
=
-\frac{2}{n}\sum_{i=1}^{n}(Y_i-c)
=
0.
$$

Следовательно,
$$
c=\overline{Y}
=
\frac{1}{n}\sum_{i=1}^{n}Y_i,
$$

$$
\boxed{
H(X)
=
\frac{1}{n}\sum_{i=1}^{n}(Y_i-\overline{Y})^2
}.
$$

**2.** $L(y, c) = |y-c|$

### 2. Абсолютная функция потерь

$$
L(y,c)=|y-c|.
$$

$$
H(X)
=
\min_{c\in\mathbb{R}}
\frac{1}{n}\sum_{i=1}^{n}|Y_i-c|.
$$

Условие минимума:
$$
0
\in
\partial_c
\left(
\frac{1}{n}\sum_{i=1}^{n}|Y_i-c|
\right)
=
\frac{1}{n}\sum_{i=1}^{n}\partial_c|Y_i-c|,
$$

где
$$
\partial_c|Y_i-c|
=
\begin{cases}
\{-1\}, & c<Y_i,\\
[-1,1], & c=Y_i,\\
\{1\}, & c>Y_i.
\end{cases}
$$

Минимум достигается на любой медиане выборки:
$$
c=\operatorname{median}(Y).
$$

Поэтому
$$
\boxed{
H(X)
=
\frac{1}{n}\sum_{i=1}^{n}
\left|Y_i-\operatorname{median}(Y)\right|
}.
$$

### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 3</span>

В случае задачи классификации рассматривается вероятностное предсказание, и мера неоднородности имеет вид
$$H(X) = \min_{\substack{P \in [0, 1]^K \\ p_1+...+p_K=1}} \dfrac{1}{n} \sum_{i=1}^n L\left(Y_i, P\right).$$
Выведите и упростите меру неоднородности, найдите оптимальные оценки вероятностей в листах для следующих функций потерь.

**1.** $L(y, P) = \sum\limits_{k=1}^K \left(p_k-I\{y=k\}\right)^2$;

### Классификация: квадратичная функция потерь

$$
H(X)
=
\min_{\substack{
p\in[0,1]^K\\
\sum_{k=1}^{K}p_k=1
}}
\frac{1}{n}\sum_{i=1}^{n}L(Y_i,p).
$$

$$
L(y,p)
=
\sum_{k=1}^{K}
\left(p_k-\mathbb{I}\{y=k\}\right)^2,
\qquad
p=
\begin{pmatrix}
p_1\\
\vdots\\
p_K
\end{pmatrix}.
$$

Обозначим
$$
I_{ik}=\mathbb{I}\{Y_i=k\}.
$$

Тогда задача минимизации имеет вид
$$
G(p)
=
\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}
(p_k-I_{ik})^2
\longrightarrow
\min_p.
$$

Функция Лагранжа:
$$
F(p,\lambda)
=
G(p)
+
\lambda\left(\sum_{k=1}^{K}p_k-1\right).
$$

Для каждого $k=1,\ldots,K$:
$$
\frac{\partial F}{\partial p_k}
=
\frac{2}{n}\sum_{i=1}^{n}(p_k-I_{ik})
+
\lambda
=
0.
$$

Отсюда
$$
p_k
-
\frac{1}{n}\sum_{i=1}^{n}I_{ik}
+
\frac{\lambda}{2}
=
0,
$$

$$
p_k
=
\frac{1}{n}\sum_{i=1}^{n}I_{ik}
-
\frac{\lambda}{2}.
$$

Используя условие нормировки, получаем
$$
1
=
\sum_{k=1}^{K}p_k
=
\frac{1}{n}
\sum_{i=1}^{n}\sum_{k=1}^{K}I_{ik}
-
\frac{\lambda K}{2}
=
1-\frac{\lambda K}{2}.
$$

Следовательно,
$$
\lambda=0,
\qquad
p_k=\frac{1}{n}\sum_{i=1}^{n}I_{ik}.
$$

Таким образом, оптимальные вероятности — доли классов в выборке:
$$
p=
\begin{pmatrix}
\displaystyle \frac{1}{n}\sum_{i=1}^{n}I_{i1}\\
\vdots\\
\displaystyle \frac{1}{n}\sum_{i=1}^{n}I_{iK}
\end{pmatrix}.
$$

Подставим их в $H(X)$. Учитывая, что
$$
I_{ik}^2=I_{ik},
\qquad
\sum_{i=1}^{n}I_{ik}=np_k,
\qquad
\sum_{k=1}^{K}p_k=1,
$$

получаем
$$
\begin{aligned}
H(X)
&=
\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}
(p_k-I_{ik})^2\\
&=
\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}
\left(p_k^2-2p_kI_{ik}+I_{ik}^2\right)\\
&=
\sum_{k=1}^{K}p_k^2
-
\frac{2}{n}\sum_{k=1}^{K}
p_k\sum_{i=1}^{n}I_{ik}
+
\frac{1}{n}\sum_{k=1}^{K}\sum_{i=1}^{n}I_{ik}\\
&=
\sum_{k=1}^{K}p_k^2
-
2\sum_{k=1}^{K}p_k^2
+
\sum_{k=1}^{K}p_k\\
&=
1-\sum_{k=1}^{K}p_k^2.
\end{aligned}
$$

$$
\boxed{
H(X)=1-\sum_{k=1}^{K}p_k^2
}
\qquad\text{— индекс Джини.}
$$

**2.** $L(y, P) = -\sum\limits_{k=1}^K I\{ y=k\}\log p_k$.

### Классификация: логарифмическая функция потерь

$$
L(y,p)
=
-\sum_{k=1}^{K}\mathbb{I}\{y=k\}\log p_k.
$$

Обозначим
$$
I_{ik}=\mathbb{I}\{Y_i=k\}.
$$

Тогда
$$
H(X)
=
\min_{\substack{
p\in[0,1]^K\\
\sum_{k=1}^{K}p_k=1
}}
\left(
-\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}
I_{ik}\log p_k
\right).
$$

$$
G(p)
=
-\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}
I_{ik}\log p_k.
$$

Функция Лагранжа:
$$
F(p,\lambda)
=
-\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}
I_{ik}\log p_k
+
\lambda\left(\sum_{k=1}^{K}p_k-1\right).
$$

Для классов, представленных в выборке:
$$
\frac{\partial F}{\partial p_k}
=
-\frac{1}{n}\sum_{i=1}^{n}\frac{I_{ik}}{p_k}
+
\lambda
=
0.
$$

Отсюда
$$
p_k
=
\frac{1}{n\lambda}\sum_{i=1}^{n}I_{ik}.
$$

По условию нормировки:
$$
1
=
\sum_{k=1}^{K}p_k
=
\frac{1}{n\lambda}
\sum_{i=1}^{n}\sum_{k=1}^{K}I_{ik}
=
\frac{1}{\lambda}.
$$

Следовательно,
$$
\lambda=1,
\qquad
p_k=\frac{1}{n}\sum_{i=1}^{n}I_{ik}.
$$

Для отсутствующих классов оптимально $p_k=0$.
Таким образом, оптимальные вероятности — доли классов в выборке.

Подставляя их в $H(X)$, получаем
$$
\begin{aligned}
H(X)
&=
-\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}
I_{ik}\log p_k\\
&=
-\sum_{k=1}^{K}\log p_k
\left(\frac{1}{n}\sum_{i=1}^{n}I_{ik}\right)\\
&=
-\sum_{k=1}^{K}p_k\log p_k.
\end{aligned}
$$

$$
\boxed{
H(X)=-\sum_{k=1}^{K}p_k\log p_k
}
\qquad\text{— энтропия распределения классов.}
$$

Используется соглашение $0\log 0=0$.
