### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 1</span>

На лекции получена формула bias-variance разложения для беггинга. Проведите эксперимент, в котором выясните, насколько уменьшается разброс (variance-компонента) беггинг-модели на 100 базовых моделях по отношению к одной базовой модели. Используйте данные [California Housing Prices](https://www.kaggle.com/datasets/camnugent/california-housing-prices), с которыми мы уже начинали работать на семинаре. Рассмотрите беггинг на следующих базовых моделях:
* решающие деревья, можно использовать вариант случайного леса,
* ридж-регрессия, используйте `BaggingRegressor(estimator=Ridge(), n_estimators=100)`.

Для решения задачи потребуется оценить корреляции предсказаний на тестовой выборке базовых моделей, входящих в состав беггинг-модели. Их можно получить с помощью поля `estimators_` у обученной беггинг-модели. Для оценки корреляций используйте [`scipy.stats.pearsonr`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.pearsonr.html) или [`numpy.corrcoef`](https://numpy.org/doc/stable/reference/generated/numpy.corrcoef.html).

Насколько уменьшается разброс в каждом случае? Для каждого случая постройте также матрицу корреляций предсказаний базовых моделей и гистограмму по ним.

```python
import copy
import random
import warnings
import os
from pathlib import Path

import numpy as np
import pandas as pd
from pandas.io.formats.style import Styler
import scipy.stats as sps

from matplotlib import pyplot as plt
import seaborn as sns
from tqdm.notebook import tqdm
from typing import Tuple, Optional, Any, Dict, List, Callable, Union, Sequence

import plotly.express as px
import kagglehub

from sklearn import datasets
from sklearn import metrics
from sklearn.metrics import accuracy_score, r2_score
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, export_graphviz
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    mean_absolute_percentage_error,
    root_mean_squared_error,
)
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV
from sklearn.base import BaseEstimator
from sklearn.ensemble import BaggingRegressor
from sklearn.linear_model import Ridge


sns.set_theme(style="whitegrid", palette="Set2", font_scale=1.2)

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

pd.set_option("display.max_columns", None)
```

```python
path = kagglehub.dataset_download("camnugent/california-housing-prices")
data = pd.read_csv(Path(path) / "housing.csv")
data.head()
```

```python
data.info()
```

```python
# Удаляем пропуски
data = data[~data.isna().any(axis=1)]

# Уникальные значения признака ocean_proximity
data["ocean_proximity"].unique()
```

```python
ocean_proximity_values = ["ISLAND", "NEAR OCEAN", "NEAR BAY", "<1H OCEAN", "INLAND"]
data["ocean_proximity_num"] = data["ocean_proximity"].apply(
    lambda value: ocean_proximity_values.index(value)
)
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt

url = (
    "https://raw.githubusercontent.com/nvkelso/"
    "natural-earth-vector/master/geojson/ne_110m_admin_1_states_provinces.geojson"
)
states = gpd.read_file(url)
california = states[states["name"] == "California"]

fig, ax = plt.subplots(figsize=(8, 10))
california.plot(ax=ax, color="#eeeeee", edgecolor="gray")

ax.scatter(
    data["longitude"],
    data["latitude"],
    s=3,
    alpha=0.3,
)
ax.set_aspect(1 / np.cos(np.deg2rad(37)))
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
plt.show()
```

```python
houses_X = data.drop(columns=["median_house_value", "ocean_proximity"])
houses_y = data["median_house_value"]
X_train, X_test, y_train, y_test = train_test_split(
    houses_X, houses_y, test_size=0.2, random_state=RANDOM_STATE
)

houses_X.shape, houses_y.shape
```

```python
n_estimators = 100
model = RandomForestRegressor(n_estimators=n_estimators, random_state=RANDOM_STATE)
model.fit(X_train, y_train)
```

```python
predictions = np.array([
    tree.predict(np.asarray(X_test))
    for tree in model.estimators_
])

correlations = np.corrcoef(predictions)
n_trees = len(model.estimators_)

pairwise_correlations = correlations[
    np.triu_indices(n_trees, k=1)
]
```

```python
predictions
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.heatmap(
    correlations,
    ax=axes[0],
    vmin=-1,
    vmax=1,
    cmap="coolwarm",
)
axes[0].set(
    title="Корреляции предсказаний деревьев",
    xlabel="Дерево",
    ylabel="Дерево",
)

axes[1].hist(pairwise_correlations, bins=30, edgecolor="black")
axes[1].set(
    title="Распределение попарных корреляций",
    xlabel="Корреляция",
    ylabel="Число пар",
)

plt.tight_layout()
plt.show()
```

```python
rho = pairwise_correlations.mean()
variance_ratio = rho + (1 - rho) / n_trees

print(f"Средняя корреляция: {rho:.4f}")
print(f"Отношение дисперсии ансамбля к дисперсии дерева: {variance_ratio:.4f}")
print(f"Уменьшение дисперсии примерно в {1 / variance_ratio:.2f} раза")
```

```python
# Проверим что теоретическое предсказание через rho сходится с практикой.
# Посчитаем среднюю дисперсию для всех деревьев и сравним ее с дисперсией всего леса.

tree_variances = predictions.var(axis=1, ddof=1)
mean_tree_variance = tree_variances.mean()

forest_predictions = predictions.mean(axis=0)
forest_variance = forest_predictions.var(ddof=1)

print(f"Средняя дисперсия деерева: {mean_tree_variance:.0f}")
print(f"Дисперсия леса: {forest_variance:.0f}")
print(f"Уменьшение дисперсии: {mean_tree_variance / forest_variance :.2f}")
```

```python
# Теперь линейная регрессия.

ridge_bagging = BaggingRegressor(estimator=Ridge(), n_estimators=n_estimators, random_state=RANDOM_STATE, n_jobs=-1)
ridge_bagging.fit(X_train, y_train)

predictions = np.array([estimator.predict(np.asarray(X_test)) for estimator in ridge_bagging.estimators_])
```

```python
correlations = np.corrcoef(predictions)

pairwise_correlations = correlations[np.triu_indices(len(ridge_bagging.estimators_), k=1)]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.heatmap(
    correlations,
    ax=axes[0],
    vmin=-1,
    vmax=1,
    cmap="coolwarm",
)
axes[0].set(
    title="Корреляции предсказаний Ridge моделей",
    xlabel="Модель",
    ylabel="Модель",
)

axes[1].hist(pairwise_correlations, bins=30, edgecolor="black")
axes[1].set(
    title="Распределение попарных корреляций",
    xlabel="Корреляция",
    ylabel="Число пар",
)

plt.tight_layout()
plt.show()
```

```python
rho = pairwise_correlations.mean()
variance_ratio = rho + (1 - rho) / n_estimators

print(f"Средняя корреляция: {rho:.4f}")
print(f"Отношение дисперсии бэггинг модели к дисперсии одной модели: {variance_ratio:.4f}")
print(f"Уменьшение дисперсии примерно в {1 / variance_ratio:.2f} раза")
```

**Вывод:**

Бутстреп производит более "случайные" модели для деревьев, чем для Ridge регрессии.  
Как результат, использование леса имеет больше смысла с точки зрения уменьшения дисперсии.  
В случае Ridge регрессии дисперсия не меняется вообще.
