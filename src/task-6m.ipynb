> Такое распределение описывает *время до наступления некоторого события* в условиях, когда события происходят независимо друг от друга. Например, таким событием может быть поступление очередного заказа таксисту или курьеру службы доставки. В таком случае $X_i$ &mdash; время ожидания $i$-го заказа после завершения предыдущего.

Иначе говоря, если приходит в среднем $\theta=15$ заказов в час, то среднее время между заказами составляет $\mathsf{E}_\theta X_i = 1/\theta = 1/15$ часа или $4$ минуты.

У вас должно получится целое множество допустимых значений, для каждого из которых вероятность ошибки 1 рода ограничена значением $\alpha$ (не обязательно равна $\alpha$).

$$

Исходя из максимизации мощности выберите константу(-ы).

**8.** Для критериев из пункта 7 постройте графики мощности в зависимости от $\theta$ для нескольких значений $n$ (не менее 5 различных значений).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gamma

theta_0 = 1
alpha = 0.05
n_values = [1, 5, 10, 30, 100]

theta = np.linspace(0.001 * theta_0, 5 * theta_0, 1500)

fig, ax = plt.subplots(figsize=(9, 5))

for n in n_values:
    threshold = gamma.ppf(alpha, a=n, scale=1 / (n * theta_0))
    power = gamma.cdf(threshold, a=n, scale=1 / (n * theta))
    ax.plot(theta, power, linewidth=2, label=f"n = {n}")

ax.axvline(theta_0, color="gray", linestyle="--", label="θ₀")
ax.axhline(alpha, color="gray", linestyle=":", label="α")

ax.set(
    xlabel="Интенсивность θ, заказов/час",
    ylabel="Мощность критерия",
    title="Критерий на увеличение интенсивности",
    ylim=(0, 1.02),
)
ax.grid(alpha=0.2)
ax.legend()

fig.tight_layout()
plt.show()
```

```python
theta = np.linspace(0.001 * theta_0, 2 * theta_0, 1500)

fig, ax = plt.subplots(figsize=(9, 5))

for n in n_values:
    threshold = gamma.isf(alpha, a=n, scale=1 / (n * theta_0))
    power = gamma.sf(threshold, a=n, scale=1 / (n * theta))
    ax.plot(theta, power, linewidth=2, label=f"n = {n}")

ax.axvline(theta_0, color="gray", linestyle="--", label="θ₀")
ax.axhline(alpha, color="gray", linestyle=":", label="α")

ax.set(
    xlabel="Интенсивность θ, заказов/час",
    ylabel="Мощность критерия",
    title="Критерий на уменьшение интенсивности",
    ylim=(0, 1.02),
)
ax.grid(alpha=0.2)
ax.legend()

fig.tight_layout()
plt.show()
```

Как меняется мощность при увеличении объёма выборки?

При увеличении объема, мощность ожидаемо насыщается быстрее (в сторону $H_1$ гипотезы). Т.е. мы становимся увереннее в критерии при меньшей разнице с нулевой гипотезой.
